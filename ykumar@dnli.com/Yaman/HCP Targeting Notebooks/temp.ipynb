```sql
select hco_id, count(distinct parent_hco_id)
from (
    select distinct *
    from (
        select distinct
            entity_vid__v as hco_id,
            parent_hco_vid__v as parent_hco_id,
            row_number() over (
                partition by entity_vid__v
                order by modified_date__v desc nulls last,
                         status_update_time__v desc nulls last
            ) as rn
        from com_raw.vod_parenthco
        where hierarchy_type__v = 'HCO_HCO'
          and entity_type__v = 'HCO'
          -- and PARENT_HCO_STATUS__V = 'A' 
          -- AND RELATIONSHIP_TYPE__V = '7356'
    )
    -- where rn=1
)
group by 1
order by 2 desc
```

```sql
SELECT DISTINCT hco_id,
                parent_hco_id
FROM (
    SELECT DISTINCT *
    FROM (
        SELECT DISTINCT
               entity_vid__v AS hco_id,
               parent_hco_vid__v AS parent_hco_id,
               ROW_NUMBER() OVER (
                   PARTITION BY entity_vid__v
                   ORDER BY modified_date__v DESC NULLS LAST,
                            status_update_time__v DESC NULLS LAST
               ) AS rn
        FROM com_raw.vod_parenthco
        WHERE hierarchy_type__v = 'HCO_HCO'
          AND entity_type__v = 'HCO'
          AND parent_hco_status__v = 'A'
    )
    WHERE rn = 1
)
-- GROUP BY 1 ORDER BY 2 DESC
```

```sql
select a.parent_hco_id, b.corporate_name__v as parent_hco_name, count(distinct a.hco_id) as child_hco_count
from (
  select distinct *
  from (
    select distinct
      entity_vid__v as hco_id,
      parent_hco_vid__v as parent_hco_id,
      row_number() over (
        partition by entity_vid__v
        order by modified_date__v desc nulls last, status_update_time__v desc nulls last
      ) as rn
    from com_raw.vod_parenthco
    where hierarchy_type__v = 'HCO_HCO'
      and entity_type__v = 'HCO'
      and parent_hco_status__v = 'A'
  )
  where rn = 1
) as a
left join com_raw.vod_hco as b on a.parent_hco_id = b.vid__v
group by 1,2 order by 3 desc
```
