```sql
WITH code_map AS (
    SELECT
        subcategory,
        REGEXP_REPLACE(UPPER(TRIM(code)), '\\.', '') AS code
    FROM codes_map_v1
    WHERE subcategory IN (
        'Developmental delay',
        'Lack of Physiological Development',
        'Hearing loss',
        'Hearing Aids'
    )
      AND code_type IN ('ICD-9-CM Diagnosis','ICD-10-CM Diagnosis')
),

-- -----------------------------------------
-- STEP 1: EXPLODE MEDICAL DX
-- -----------------------------------------
medical_dx AS (
    SELECT
        m.patient_id,
        REGEXP_REPLACE(UPPER(TRIM(dx_token)), '\\.', '') AS dx_code
    FROM com_edp_prd.com_raw.kom_medical_events m
    LATERAL VIEW OUTER explode(split(coalesce(m.diagnosis_codes,''), '\\|')) dx AS dx_token
    WHERE m.service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
),

-- -----------------------------------------
-- STEP 2: MAP DX → SUBCATEGORY
-- -----------------------------------------
mapped_dx AS (
    SELECT
        md.patient_id,
        cm.subcategory
    FROM medical_dx md
    JOIN code_map cm
        ON md.dx_code = cm.code
),

-- -----------------------------------------
-- STEP 3: IDENTIFY TARGET PATIENTS
-- -----------------------------------------
patient_flags AS (
    SELECT
        patient_id,
        MAX(CASE WHEN subcategory = 'Developmental delay' THEN 1 ELSE 0 END) AS dev_delay,
        MAX(CASE WHEN subcategory = 'Lack of Physiological Development' THEN 1 ELSE 0 END) AS phys_dev,
        MAX(CASE WHEN subcategory IN ('Hearing loss','Hearing Aids') THEN 1 ELSE 0 END) AS hearing_flag
    FROM mapped_dx
    GROUP BY patient_id
),

target_patients AS (
    SELECT patient_id
    FROM patient_flags
    WHERE dev_delay = 1
      AND phys_dev = 1
      AND hearing_flag = 1
),

-- -----------------------------------------
-- STEP 4: ALL DX FOR TARGET PATIENTS
-- -----------------------------------------
final_dx AS (
    SELECT
        md.patient_id,
        md.dx_code
    FROM medical_dx md
    JOIN target_patients t
        ON md.patient_id = t.patient_id
),

-- -----------------------------------------
-- STEP 5: PHARMACY NDC DATA
-- -----------------------------------------
pharmacy_ndc AS (
    SELECT
        p.patient_id,
        p.NDC11,
        COUNT(*) AS claims
    FROM com_edp_prd.com_raw.kom_pharmacy_events p
    JOIN target_patients t
        ON p.patient_id = t.patient_id
    WHERE p.fill_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
      AND p.transaction_status IN ('Final','Standalone')
      AND p.transaction_result = 'Paid'
    GROUP BY p.patient_id, p.NDC11
),

-- -----------------------------------------
-- STEP 6: MEDICAL DRUGS (HCPCS)
-- -----------------------------------------
medical_drugs AS (
    SELECT
        m.patient_id,
        m.procedure_code AS hcpcs_code,
        COUNT(*) AS claims
    FROM com_edp_prd.com_raw.kom_medical_events m
    JOIN target_patients t
        ON m.patient_id = t.patient_id
    WHERE m.service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
      AND m.procedure_code IS NOT NULL
    GROUP BY m.patient_id, m.procedure_code
)

-- =========================================
-- FINAL OUTPUTS
-- =========================================

-- 1️⃣ TOTAL PATIENT COUNT
SELECT 'Total Patients' AS metric, COUNT(DISTINCT patient_id) AS value
FROM target_patients

UNION ALL

-- 2️⃣ TOP DX CODES
SELECT
    CONCAT('DX: ', dx_code) AS metric,
    COUNT(DISTINCT patient_id) AS value
FROM final_dx
GROUP BY dx_code

UNION ALL

-- 3️⃣ TOP NDCs
SELECT
    CONCAT('NDC: ', NDC11) AS metric,
    COUNT(DISTINCT patient_id) AS value
FROM pharmacy_ndc
GROUP BY NDC11

UNION ALL

-- 4️⃣ TOP HCPCS (MEDICAL DRUGS)
SELECT
    CONCAT('HCPCS: ', hcpcs_code) AS metric,
    COUNT(DISTINCT patient_id) AS value
FROM medical_drugs
GROUP BY hcpcs_code

ORDER BY value DESC;
```

```sql
-- =============================================================================
-- Continuous enrollment check for ONLY: index_date - 6 months  to  index_date + 6 months
-- Business rules :
--   • merge overlapping/adjacent enrollment rows (<= 1 day gap allowed)
--   • pick the merged block that covers index_date
--   • flag = 1 if that block fully covers [index-6mo, index+6mo]
-- Output: distinct patient_id with continuous coverage in +/- 6 months window
-- =============================================================================

create or replace temporary view continuous_enrollment_patients_pm6m as
WITH idx AS (
  SELECT
    patient_id,
    first_elaprase_treatment_date AS index_date,

    -- window bounds: +/- 6 months
    add_months(first_elaprase_treatment_date, -6) AS pre_6m_start,
    add_months(first_elaprase_treatment_date,  6) AS post_6m_end

  FROM first_elaprase_treatment
),

-- only enrollment rows that could affect the +/- 6 month window
enr AS (
  SELECT
    e.patient_id,
    e.start_date,
    e.end_date
  FROM com_raw.kom_patient_enrollment e
  JOIN idx i
    ON e.patient_id = i.patient_id
  WHERE e.end_date  >= i.pre_6m_start
    AND e.start_date <= i.post_6m_end
),

-- detect gaps (gaps & islands) with 1-day adjacency allowed (same rule as above)
ordered AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    CASE
      WHEN lag(end_date) OVER (
             PARTITION BY patient_id ORDER BY start_date, end_date
           ) IS NULL
      THEN 1
      WHEN start_date <= date_add(
             lag(end_date) OVER (PARTITION BY patient_id ORDER BY start_date, end_date),
             1
           )
      THEN 0
      ELSE 1
    END AS new_block
  FROM enr
),

blocks AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    sum(new_block) OVER (
      PARTITION BY patient_id ORDER BY start_date, end_date
      ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS block_id
  FROM ordered
),

merged AS (
  SELECT
    patient_id,
    block_id,
    min(start_date) AS block_start,
    max(end_date)   AS block_end
  FROM blocks
  GROUP BY patient_id, block_id
),

-- get the block that covers index_date
covering AS (
  SELECT
    i.patient_id,
    i.index_date,
    i.pre_6m_start,
    i.post_6m_end,
    m.block_start,
    m.block_end
  FROM idx i
  LEFT JOIN merged m
    ON i.patient_id = m.patient_id
   AND i.index_date BETWEEN m.block_start AND m.block_end
),

flags AS (
  SELECT
    patient_id,
    index_date,
    CASE
      WHEN block_start <= pre_6m_start
       AND block_end   >= post_6m_end
      THEN 1 ELSE 0
    END AS cont_pm_6m_flag
  FROM covering
)

SELECT DISTINCT patient_id
FROM flags
WHERE cont_pm_6m_flag = 1;
```

```sql
CREATE OR REPLACE TABLE cmpa_insights_internal_schema.comorbidity_claims_occurence_data  AS
WITH code_map AS (
  SELECT DISTINCT
    new_bucket,
    subcategory,
    UPPER(TRIM(code)) AS code,
    code_type,
    CASE
      WHEN code_type IN ('ICD-9-CM Diagnosis', 'ICD-10-CM Diagnosis') THEN 'DX'
      WHEN code_type IN ('CPT', 'HCPCS Level II') THEN 'PROC'
      ELSE 'OTHER'
    END AS match_domain
  FROM codes_map_v1
  WHERE subcategory in 
  (
      'ADHD',
      'Anxiety Disorders',
      'Behavioral Issues',
      'Developmental delay',
      'Hydrocephalus',
      'Lack of Physiological Development',
      'Mood disorders',
      'Seizures',
      'Carpal tunnel syndrome',
      'Hearing Aids',
      'Hearing loss',
      'Retinopathy',
      'Sleep related disorders',
      'Spinal Disorders'
)
),

dx_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'DX'
),

proc_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'PROC'
),

medical_base AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
),

medical_dx_exploded AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type,
    REGEXP_REPLACE(UPPER(TRIM(dx_token)), '\\.', '') AS dx_code
  FROM medical_base
  LATERAL VIEW OUTER explode(split(coalesce(diagnosis_codes,''), '\\|')) dx AS dx_token
),

medical_dx AS (
  SELECT
    me.patient_id,
    me.medical_event_id AS claim_id,
    me.service_date      AS fill_date,
    'MEDICAL'            AS source,
    me.diagnosis_codes   AS diagnosis_codes,
    me.procedure_code    AS procedure_code,
    me.procedure_code_type AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM medical_dx_exploded me
  JOIN dx_map dm
    ON me.dx_code = REGEXP_REPLACE(dm.code, '\\.', '')
  WHERE me.dx_code IS NOT NULL AND me.dx_code <> ''
),

medical_proc AS (
  SELECT
    mb.patient_id,
    mb.medical_event_id AS claim_id,
    mb.service_date      AS fill_date,
    'MEDICAL'            AS source,
    mb.diagnosis_codes   AS diagnosis_codes,
    mb.procedure_code    AS procedure_code,
    mb.procedure_code_type AS procedure_code_type,
    'PROC'               AS types,
    pm.code              AS matched_code,
    pm.new_bucket        AS matched_new_bucket,
    pm.subcategory       AS matched_subcategory
  FROM medical_base mb
  JOIN proc_map pm
    ON UPPER(TRIM(mb.procedure_code)) = pm.code
),

pharmacy_base AS (
  SELECT
    patient_id,
    pharmacy_event_id,
    fill_date,
    diagnosis_code
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
    AND transaction_status IN ('Final', 'Standalone')
    AND transaction_result = 'Paid'
),

pharmacy_dx AS (
  SELECT
    pb.patient_id,
    pb.pharmacy_event_id AS claim_id,
    pb.fill_date         AS fill_date,
    'PHARMACY'           AS source,
    pb.diagnosis_code    AS diagnosis_codes,
    CAST(NULL AS STRING) AS procedure_code,
    CAST(NULL AS STRING) AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM pharmacy_base pb
  JOIN dx_map dm
    ON REGEXP_REPLACE(UPPER(TRIM(pb.diagnosis_code)), '\\.', '') = REGEXP_REPLACE(dm.code, '\\.', '')
),

final_union AS (
  SELECT * FROM medical_dx
  UNION ALL
  SELECT * FROM medical_proc
  UNION ALL
  SELECT * FROM pharmacy_dx
),

qc_check AS (
  SELECT
    t.*,
    CASE
      WHEN t.matched_code IS NULL OR TRIM(t.matched_code) = '' THEN 0
      WHEN LOWER(t.diagnosis_codes) LIKE '%|' || LOWER(t.matched_code) || '|%'
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) = LOWER(t.matched_code)
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) LIKE '%' || LOWER(t.matched_code) || '%'
      THEN 1
      ELSE 0
    END AS matched_code_present
  FROM final_union t
),

-- ✅ 1-claim rule: patient must have >= 1 DISTINCT fill_date for the SAME matched_code
qualifying_patient_code AS (
  SELECT
    patient_id,
    matched_code
  FROM qc_check
  WHERE matched_code_present = 1
  GROUP BY patient_id, matched_code
  HAVING COUNT(DISTINCT fill_date) >= 1
),

filtered_claims AS (
  SELECT q.*
  FROM qc_check q
  JOIN qualifying_patient_code k
    ON q.patient_id = k.patient_id
   AND q.matched_code = k.matched_code
  WHERE q.matched_code_present = 1
),

patient_age AS (
  SELECT
    a.*,
    (year(current_date) - year(b.patient_yob)) AS age
  FROM filtered_claims a
  LEFT JOIN (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
      SELECT
        PATIENT_ID AS patient_id,
        PATIENT_YOB AS patient_yob,
        row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
      FROM com_raw.kom_patient_demographics
    )
    WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
)

SELECT * FROM patient_age where patient_id in (select distinct patient_id from eligible_patient_cohort);
```

```sql
%python

# -----------------------------
# Step 1: Load data
# -----------------------------
df = spark.table("cmpa_insights_internal_schema.comorbidity_claims_occurence_data") \
          .select("patient_id", "matched_subcategory") \
          .dropDuplicates() \
          .toPandas()

# -----------------------------
# Step 2: Create patient-subcategory matrix
# -----------------------------
pivot = pd.crosstab(df['patient_id'], df['matched_subcategory'])

# Convert to binary (presence/absence)
pivot = (pivot > 0).astype(int)

# -----------------------------
# Step 3: Create combination string per patient
# -----------------------------
pivot['combination'] = pivot.apply(
    lambda row: '|'.join(sorted(row.index[row == 1])),
    axis=1
)

# -----------------------------
# Step 4: Count patients per combination
# -----------------------------
combo_counts = pivot.groupby('combination') \
                    .size() \
                    .reset_index(name='patient_count') \
                    .sort_values(by='patient_count', ascending=False)

# -----------------------------
# Step 5: Keep Top N combinations (for visualization clarity)
# -----------------------------
TOP_N = 20
top_combos = combo_counts.head(TOP_N)

# -----------------------------
# Step 6: Convert combinations back to matrix form
# -----------------------------
def combo_to_vector(combo, categories):
    return [1 if cat in combo.split('|') else 0 for cat in categories]

categories = list(pivot.columns)
categories.remove('combination')

matrix = pd.DataFrame(
    [combo_to_vector(c, categories) for c in top_combos['combination']],
    columns=categories
)

matrix['patient_count'] = top_combos['patient_count'].values

# -----------------------------
# Step 7: Sort by patient count
# -----------------------------
matrix = matrix.sort_values(by='patient_count', ascending=False)

# -----------------------------
# Step 8: Plot block heatmap (patient combinations)
# -----------------------------
plt.figure(figsize=(14, 8))

# Only comorbidity presence (exclude patient_count)
plot_data = matrix[categories]

plt.imshow(plot_data, aspect='auto', cmap='Greens')

# Labels
plt.xticks(range(len(categories)), categories, rotation=90)
plt.yticks(range(len(matrix)), matrix['patient_count'])

plt.xlabel("Comorbidities")
plt.ylabel("Patient Count (per combination)")

# Title
plt.title("Top Patient-Level Comorbidity Combinations", pad=20)

# Annotate blocks
for i in range(plot_data.shape[0]):
    for j in range(plot_data.shape[1]):
        if plot_data.iloc[i, j] == 1:
            plt.text(j, i, '■', ha='center', va='center', fontsize=10)

plt.colorbar(label="Presence (1=Yes, 0=No)")
plt.tight_layout()
plt.show()
```

```sql
-- =============================================================================
-- Combined table with:
--  • Eligible cohort denominators
--  • Whole number percentages
--  • "Overall" instead of NULL for bucket totals
-- =============================================================================

WITH
-- -----------------------------------------
-- 1) Eligible cohort denominators
-- -----------------------------------------
eligible_total AS (
  SELECT COUNT(DISTINCT patient_id) AS total_n
  FROM eligible_patient_cohort
),

eligible_patients_with_age AS (
  SELECT
      a.patient_id,
      (year(current_date) - year(b.patient_yob)) AS age
  FROM eligible_patient_cohort a
  LEFT JOIN (
      SELECT patient_id, patient_yob
      FROM (
          SELECT
              PATIENT_ID AS patient_id,
              PATIENT_YOB AS patient_yob,
              row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
          FROM com_raw.kom_patient_demographics
          WHERE PATIENT_YOB IS NOT NULL
      ) x
      WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
),

eligible_age_buckets AS (
  SELECT
      patient_id,
      CASE
          WHEN age BETWEEN 0 AND 5 THEN '0-5'
          WHEN age BETWEEN 6 AND 11 THEN '6-11'
          WHEN age BETWEEN 12 AND 17 THEN '12-17'
          WHEN age >= 18 THEN '18+'
          ELSE NULL
      END AS age_band
  FROM eligible_patients_with_age
),

eligible_age_denoms AS (
  SELECT
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END) AS n_0_5,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END) AS n_6_11,
      COUNT(DISTINCT CASE WHEN age_band = '12-17' THEN patient_id END) AS n_12_17,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END) AS n_18_plus
  FROM eligible_age_buckets
),

denom AS (
  SELECT *
  FROM eligible_total
  CROSS JOIN eligible_age_denoms
),

-- -----------------------------------------
-- 2) Matched cohort (numerators)
-- -----------------------------------------
cohort AS (
  SELECT *
  FROM cmpa_insights_internal_schema.comorbidity_claims
  WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
    AND matched_code_present = 1
),

base AS (
  SELECT DISTINCT
      patient_id,
      matched_new_bucket,
      matched_subcategory,
      CASE
          WHEN age BETWEEN 0 AND 5 THEN '0-5'
          WHEN age BETWEEN 6 AND 11 THEN '6-11'
          WHEN age BETWEEN 12 AND 17 THEN '12-17'
          WHEN age >= 18 THEN '18+'
          ELSE NULL
      END AS age_band
  FROM cohort
),

bucket_totals AS (
  SELECT
      matched_new_bucket,
      'Overall' AS matched_subcategory,

      COUNT(DISTINCT patient_id) AS overall_n,
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END) AS age_0_5_n,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END) AS age_6_11_n,
      COUNT(DISTINCT CASE WHEN age_band = '12-17' THEN patient_id END) AS age_12_17_n,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END) AS age_18_plus_n
  FROM base
  GROUP BY matched_new_bucket
),

bucket_counts AS (
  SELECT
      matched_new_bucket,
      matched_subcategory,

      COUNT(DISTINCT patient_id) AS overall_n,
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END) AS age_0_5_n,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END) AS age_6_11_n,
      COUNT(DISTINCT CASE WHEN age_band = '12-17' THEN patient_id END) AS age_12_17_n,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END) AS age_18_plus_n
  FROM base
  GROUP BY matched_new_bucket, matched_subcategory
),

final_counts AS (
  SELECT * FROM bucket_totals
  UNION ALL
  SELECT * FROM bucket_counts
)

-- -----------------------------------------
-- 3) Final Output
-- -----------------------------------------
SELECT
    f.matched_new_bucket  AS `Clinical Complication`,
    f.matched_subcategory AS `complication subcategory`,

    CONCAT(
      f.overall_n, ' (',
      ROUND(100.0 * f.overall_n / NULLIF(d.total_n, 0), 0),
      ')'
    ) AS `#Patients`,

    CONCAT(
      f.age_0_5_n, ' (',
      ROUND(100.0 * f.age_0_5_n / NULLIF(d.n_0_5, 0), 0),
      ')'
    ) AS `0-5 years`,

    CONCAT(
      f.age_6_11_n, ' (',
      ROUND(100.0 * f.age_6_11_n / NULLIF(d.n_6_11, 0), 0),
      ')'
    ) AS `6-11 years`,

    CONCAT(
      f.age_12_17_n, ' (',
      ROUND(100.0 * f.age_12_17_n / NULLIF(d.n_12_17, 0), 0),
      ')'
    ) AS `12-17 years`,

    CONCAT(
      f.age_18_plus_n, ' (',
      ROUND(100.0 * f.age_18_plus_n / NULLIF(d.n_18_plus, 0), 0),
      ')'
    ) AS `18+ years`

FROM final_counts f
CROSS JOIN denom d
ORDER BY
    f.matched_new_bucket,
    CASE WHEN f.matched_subcategory = 'Overall' THEN 0 ELSE 1 END,
    f.matched_subcategory;
```

```sql
-- A: Avg and Median age at first incident per comorbidity
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),
eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),
first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        matched_subcategory,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket, matched_subcategory
),
first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        f.matched_subcategory,
        (year(f.first_incident_date) - year(p.patient_yob)) AS age_at_first_incident
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
)
SELECT
    matched_new_bucket                                    AS `Clinical Complication`,
    matched_subcategory                                   AS `Complication Subcategory`,
    COUNT(DISTINCT patient_id)                            AS `Patient Count`,
    ROUND(AVG(age_at_first_incident), 1)                  AS `Avg Age at First Incident`,
    ROUND(PERCENTILE(age_at_first_incident, 0.5), 1)      AS `Median Age at First Incident`
FROM first_incident_with_age
WHERE age_at_first_incident IS NOT NULL
GROUP BY matched_new_bucket, matched_subcategory
ORDER BY matched_new_bucket, matched_subcategory;
```

```sql
-- Output B: Age distribution at first incident (bell curve data)
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),
eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),
first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),
first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        (year(f.first_incident_date) - year(p.patient_yob)) AS age_at_first_incident
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
),
bucket_totals AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_in_bucket
    FROM first_incident_with_age
    WHERE age_at_first_incident IS NOT NULL
    GROUP BY matched_new_bucket
),
age_counts AS (
    SELECT
        matched_new_bucket,
        age_at_first_incident AS age,
        COUNT(DISTINCT patient_id) AS n_patients
    FROM first_incident_with_age
    WHERE age_at_first_incident IS NOT NULL
    GROUP BY matched_new_bucket, age_at_first_incident
)
SELECT
    a.matched_new_bucket                                        AS `Clinical Complication`,
    a.age                                                       AS `Age at First Incident`,
    a.n_patients                                                AS `# Patients`,
    ROUND(100.0 * a.n_patients / NULLIF(b.total_patients_in_bucket, 0), 1) AS `% of Complication Population`
FROM age_counts a
JOIN bucket_totals b ON a.matched_new_bucket = b.matched_new_bucket
ORDER BY a.matched_new_bucket, a.age;
```

```sql
-- =============================================================================
-- Continuous enrollment check for ONLY: index_date - 6 months  to  index_date + 6 months
-- Business rules SAME as your current logic:
--   • merge overlapping/adjacent enrollment rows (<= 1 day gap allowed)
--   • pick the merged block that covers index_date
--   • flag = 1 if that block fully covers [index-6mo, index+6mo]
-- Output: distinct patient_id with continuous coverage in +/- 6 months window
-- =============================================================================

create or replace temporary view continuous_enrollment_patients_pm6m as
WITH idx AS (
  SELECT
    patient_id,
    first_elaprase_treatment_date AS index_date,

    -- window bounds: +/- 6 months
    add_months(first_elaprase_treatment_date, -6) AS pre_6m_start,
    add_months(first_elaprase_treatment_date,  6) AS post_6m_end

  FROM first_elaprase_treatment
),

-- only enrollment rows that could affect the +/- 6 month window
enr AS (
  SELECT
    e.patient_id,
    e.start_date,
    e.end_date
  FROM com_raw.kom_patient_enrollment e
  JOIN idx i
    ON e.patient_id = i.patient_id
  WHERE e.end_date  >= i.pre_6m_start
    AND e.start_date <= i.post_6m_end
),

-- detect gaps (gaps & islands) with 1-day adjacency allowed (same rule as above)
ordered AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    CASE
      WHEN lag(end_date) OVER (
             PARTITION BY patient_id ORDER BY start_date, end_date
           ) IS NULL
      THEN 1
      WHEN start_date <= date_add(
             lag(end_date) OVER (PARTITION BY patient_id ORDER BY start_date, end_date),
             1
           )
      THEN 0
      ELSE 1
    END AS new_block
  FROM enr
),

blocks AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    sum(new_block) OVER (
      PARTITION BY patient_id ORDER BY start_date, end_date
      ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS block_id
  FROM ordered
),

merged AS (
  SELECT
    patient_id,
    block_id,
    min(start_date) AS block_start,
    max(end_date)   AS block_end
  FROM blocks
  GROUP BY patient_id, block_id
),

-- get the block that covers index_date
covering AS (
  SELECT
    i.patient_id,
    i.index_date,
    i.pre_6m_start,
    i.post_6m_end,
    m.block_start,
    m.block_end
  FROM idx i
  LEFT JOIN merged m
    ON i.patient_id = m.patient_id
   AND i.index_date BETWEEN m.block_start AND m.block_end
),

flags AS (
  SELECT
    patient_id,
    index_date,
    CASE
      WHEN block_start <= pre_6m_start
       AND block_end   >= post_6m_end
      THEN 1 ELSE 0
    END AS cont_pm_6m_flag
  FROM covering
)

SELECT DISTINCT patient_id
FROM flags
WHERE cont_pm_6m_flag = 1;
```

```sql
/* ============================================================
   Cohort funnel + view creation (Elaprase exposure → dx-confirm → male)
   Window: 2015-01-01 to 2025-12-31
   All base tables are prefixed with: com_edp_prd.com_raw.
   ============================================================ */

-- ------------------------------------------------------------
-- 1) Create Elaprase exposure view (medical NDC + pharmacy NDC + medical J-code)
-- ------------------------------------------------------------
WITH
medical_ndc AS (
  SELECT
    patient_id,
    rendering_npi AS npi,
    ndc_code AS code,
    event_id,
    service_date AS fill_date,
    place_of_service,
    plan_id,
    'kom_medical_events' AS source_table
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN DATE '2015-01-01' AND DATE '2025-12-31'
    AND ndc_code IN ('54092070001','540920700')
),
pharmacy_ndc AS (
  SELECT
    patient_id,
    prescriber_npi AS npi,
    ndc_code AS code,
    event_id,
    fill_date,
    NULL AS place_of_service,
    plan_id,
    'kom_pharmacy_events' AS source_table
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN DATE '2015-01-01' AND DATE '2025-12-31'
    AND ndc_code IN ('54092070001','540920700')
    AND transaction_status IN ('Final','Standalone')
    AND transaction_result = 'Paid'
),
medical_jcode AS (
  SELECT
    patient_id,
    rendering_npi AS npi,
    hcpcs_code AS code,
    event_id,
    service_date AS fill_date,
    place_of_service,
    plan_id,
    'kom_medical_events' AS source_table
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN DATE '2015-01-01' AND DATE '2025-12-31'
    AND hcpcs_code = 'J1743'
)
SELECT * FROM medical_ndc
UNION ALL
SELECT * FROM pharmacy_ndc
UNION ALL
SELECT * FROM medical_jcode
;

-- ------------------------------------------------------------
-- 2) Create MPS II diagnosis view (E76.1 often stored as E761)
-- ------------------------------------------------------------
CREATE OR REPLACE VIEW diagnosis_closed_claims AS
WITH
medical_dx AS (
  SELECT
    patient_id,
    rendering_npi AS npi,
    diagnosis_codes AS diagnosis_field,
    event_id,
    service_date AS fill_date,
    place_of_service,
    plan_id,
    'kom_medical_events' AS source_table
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN DATE '2015-01-01' AND DATE '2025-12-31'
    AND diagnosis_codes LIKE '%E761%'
),
pharmacy_dx AS (
  SELECT
    patient_id,
    prescriber_npi AS npi,
    diagnosis_code AS diagnosis_field,
    event_id,
    fill_date,
    NULL AS place_of_service,
    plan_id,
    'kom_pharmacy_events' AS source_table
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN DATE '2015-01-01' AND DATE '2025-12-31'
    AND diagnosis_code = 'E761'
    AND transaction_status IN ('Final','Standalone')
    AND transaction_result = 'Paid'
)
SELECT * FROM medical_dx
UNION ALL
SELECT * FROM pharmacy_dx
;

-- ------------------------------------------------------------
-- 3) Funnel counts as a single output table
-- ------------------------------------------------------------
WITH
step1_treated AS (
  SELECT DISTINCT patient_id
  FROM elaprase_closed_claims
),

treatment_first AS (
  SELECT
    patient_id,
    MIN(fill_date) AS index_date
  FROM elaprase_closed_claims
  GROUP BY patient_id
),

dx_all AS (
  SELECT
    patient_id,
    fill_date AS dx_date
  FROM diagnosis_closed_claims
),

dx_in_window AS (
  SELECT
    t.patient_id,
    d.dx_date
  FROM treatment_first t
  JOIN dx_all d
    ON t.patient_id = d.patient_id
  WHERE d.dx_date BETWEEN DATEADD(day, -365, t.index_date) AND t.index_date
),

step2_dx_any AS (
  SELECT DISTINCT patient_id
  FROM dx_in_window
),

dx_distinct_dates AS (
  SELECT DISTINCT patient_id, dx_date
  FROM dx_in_window
),

dx_with_lag AS (
  SELECT
    patient_id,
    dx_date,
    LAG(dx_date) OVER (PARTITION BY patient_id ORDER BY dx_date) AS prev_dx_date
  FROM dx_distinct_dates
),

step3_dx_confirmed AS (
  SELECT DISTINCT patient_id
  FROM dx_with_lag
  WHERE prev_dx_date IS NOT NULL
    AND DATEDIFF(day, prev_dx_date, dx_date) <= 90
),

step4_male_confirmed AS (
  SELECT DISTINCT s3.patient_id
  FROM step3_dx_confirmed s3
  JOIN com_edp_prd.com_raw.kom_patient_demographics demo
    ON s3.patient_id = demo.patient_id
  WHERE demo.patient_gender = 'M'
),

counts AS (
  SELECT 'Step 1: ≥1 Elaprase (idursulfase) claim' AS funnel_step, COUNT(DISTINCT patient_id) AS n
  FROM step1_treated
  UNION ALL
  SELECT 'Step 2: ≥1 MPS II dx in 365d pre-index', COUNT(DISTINCT patient_id)
  FROM step2_dx_any
  UNION ALL
  SELECT 'Step 3: ≥2 MPS II dx dates ≤90d apart (pre-index)', COUNT(DISTINCT patient_id)
  FROM step3_dx_confirmed
  UNION ALL
  SELECT 'Step 4: Male only', COUNT(DISTINCT patient_id)
  FROM step4_male_confirmed
),

step1_n AS (
  SELECT n AS n_step1
  FROM counts
  WHERE funnel_step = 'Step 1: ≥1 Elaprase (idursulfase) claim'
)

SELECT
  c.funnel_step,
  c.n AS patient_count,
  ROUND(100.0 * c.n / NULLIF(s.n_step1, 0), 1) AS pct_of_step1
FROM counts c
CROSS JOIN step1_n s
ORDER BY
  CASE c.funnel_step
    WHEN 'Step 1: ≥1 Elaprase (idursulfase) claim' THEN 1
    WHEN 'Step 2: ≥1 MPS II dx in 365d pre-index' THEN 2
    WHEN 'Step 3: ≥2 MPS II dx dates ≤90d apart (pre-index)' THEN 3
    WHEN 'Step 4: Male only' THEN 4
    ELSE 99
  END;
```
