```sql
WITH hco_master AS (
    SELECT DISTINCT
        LOWER(TRIM(hco_name)) AS hco_name,
        LOWER(TRIM(hco_address)) AS hco_address,
        hco_zip,
        hco_city,
        hco_state
    FROM com_edp_prd.cmpa_insights_internal_schema.reference_file_pooja_1703
)
SELECT 
    hco_name,
    COUNT(DISTINCT hco_address) AS address_count
FROM hco_master
GROUP BY 1
HAVING COUNT(DISTINCT hco_address) > 1
```

```sql
WITH hco_master AS (
    SELECT DISTINCT
        LOWER(TRIM(hco_name)) AS hco_name,
        LOWER(TRIM(hco_address)) AS hco_address,
        hco_zip,
        hco_city,
        hco_state
    FROM com_edp_prd.cmpa_insights_internal_schema.reference_file_pooja_1703
)
SELECT 
    hco_address,
    hco_zip,
    COUNT(DISTINCT hco_name) AS name_count
FROM hco_master
GROUP BY 1,2
HAVING COUNT(DISTINCT hco_name) > 1 ;
```

```sql
WITH hco_master AS (
    SELECT DISTINCT
        LOWER(TRIM(hco_name)) AS hco_name,
        LOWER(TRIM(hco_address)) AS hco_address,
        hco_zip,
        hco_city,
        hco_state
    FROM com_edp_prd.cmpa_insights_internal_schema.reference_file_pooja_1703
)
SELECT 
    a.hco_name AS hco_name_1,
    b.hco_name AS hco_name_2,
    a.hco_address,
    levenshtein(a.hco_name, b.hco_name) AS distance
FROM hco_master a
JOIN hco_master b
  ON a.hco_address = b.hco_address
 AND a.hco_name < b.hco_name
WHERE levenshtein(a.hco_name, b.hco_name) <= 3
```
