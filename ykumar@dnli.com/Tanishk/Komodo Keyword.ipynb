```sql
CREATE OR REPLACE TEMP VIEW hco_universe AS
SELECT DISTINCT
  p.npi                 AS hco_npi,
  p.organization_name   AS hco_name,
  p.provider_address    AS hco_address,
  p.provider_city       AS hco_city,
  p.provider_state      AS hco_state,
  p.provider_zip        AS hco_zip
FROM com_raw.kom_providers p
WHERE p.provider_type = 'ORGANIZATION'
  AND p.npi IN (
    SELECT DISTINCT hco_npi
    FROM com_edp_prd.cmpa_insights_internal_schema.reference_file_0109
  );
```

```sql
CREATE OR REPLACE TEMP VIEW hco_universe_clean AS
SELECT
  *,
  trim(
    regexp_replace(
      lower(regexp_replace(hco_name, '[^a-zA-Z0-9\\s]', '')),
      '\\s+',
      ' '
    )
  ) AS hco_name_norm
FROM hco_universe;
```

```sql
CREATE OR REPLACE TEMP VIEW keyword_dict_clean AS
SELECT
  hco_group_name,
  trim(
    regexp_replace(
      lower(regexp_replace(trim(k), '[^a-zA-Z0-9\\s]', '')),
      '\\s+',
      ' '
    )
  ) AS keyword_norm
FROM keyword_source
LATERAL VIEW explode(split(keyword_string, '\\s*;\\s*')) e AS k
WHERE trim(k) <> '';
```

```sql
CREATE OR REPLACE TEMP VIEW hco_tagging_matches AS
SELECT DISTINCT
  h.hco_npi,
  h.hco_name,
  k.hco_group_name AS tagged_group,
  k.keyword_norm   AS matched_keyword,
  h.hco_address,
  h.hco_city,
  h.hco_state,
  h.hco_zip
  
FROM hco_universe_clean h
JOIN keyword_dict_clean k
  ON h.hco_name_norm LIKE concat('%', k.keyword_norm, '%');
```

```sql
SELECT *
FROM hco_tagging_matches
ORDER BY tagged_group, hco_name;
```
