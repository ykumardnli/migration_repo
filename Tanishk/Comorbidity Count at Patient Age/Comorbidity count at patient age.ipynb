```sql
WITH pool AS (
  SELECT
    patient_id,
    year(patient_yob) AS birth_year
  FROM com_edp_prd.cmpa_insights_internal_schema.patient360_master
),

events AS (
  SELECT
    e.patient_id,
    to_date(e.service_date) AS service_date_dt,
    e.diagnosis_codes,
    p.birth_year
  FROM com_edp_prd.com_raw.kom_medical_events e
  JOIN pool p
    ON e.patient_id = p.patient_id
  WHERE e.diagnosis_codes IS NOT NULL
    AND to_date(e.service_date) IS NOT NULL
),

dx AS (
  -- explode diagnosis_codes into one code per row
  SELECT
    patient_id,
    service_date_dt,
    (year(service_date_dt) - birth_year) AS age_at_dx,
    upper(trim(dx_code)) AS dx_code
  FROM events
  LATERAL VIEW explode(
    split(
      regexp_replace(diagnosis_codes, '[,\\s]', ''),   -- remove commas/spaces
      '\\|'
    )
  ) s AS dx_code
  WHERE dx_code IS NOT NULL AND trim(dx_code) <> ''
),

mapped AS (
  SELECT
    d.patient_id,
    d.age_at_dx,
    m.symptom,
    d.dx_code
  FROM dx d
  JOIN symptom_code_map m
    ON d.dx_code = m.code
)

SELECT
  patient_id,
  age_at_dx AS age,
  symptom,
  COUNT(DISTINCT dx_code) AS distinct_code_count
FROM mapped
GROUP BY patient_id, age_at_dx, symptom
ORDER BY patient_id, age_at_dx, symptom;
```

```sql
WITH pool AS (
  SELECT patient_id, year(patient_yob) AS birth_year
  FROM com_edp_prd.cmpa_insights_internal_schema.patient360_master
),

events AS (
  SELECT
    e.patient_id,
    to_date(e.service_date) AS service_date_dt,
    e.diagnosis_codes,
    p.birth_year
  FROM com_edp_prd.com_raw.kom_medical_events e
  JOIN pool p ON e.patient_id = p.patient_id
  WHERE e.diagnosis_codes IS NOT NULL
    AND to_date(e.service_date) IS NOT NULL
),

dx AS (
  SELECT
    patient_id,
    service_date_dt,
    diagnosis_codes,                               -- keep raw string
    (year(service_date_dt) - birth_year) AS age_at_dx,
    upper(trim(dx_code)) AS dx_code
  FROM events
  LATERAL VIEW explode(
    split(regexp_replace(diagnosis_codes, '[,\\s]', ''), '\\|')
  ) s AS dx_code
  WHERE dx_code IS NOT NULL AND trim(dx_code) <> ''
),

mapped_detail AS (
  SELECT
    d.patient_id,
    d.age_at_dx,
    d.service_date_dt,
    m.symptom,
    d.dx_code,
    d.diagnosis_codes
  FROM dx d
  JOIN symptom_code_map m
    ON d.dx_code = m.code
)

SELECT *
FROM mapped_detail
ORDER BY patient_id, age_at_dx, symptom, dx_code, service_date_dt;
```
