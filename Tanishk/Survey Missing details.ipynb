```sql
with survey_spine as (
  select
    a.*,
    b.npi__v as npi,
    a.name__v as survey_name
  from com_intgr.survey_target a
  left join com_intgr.customer b
    on a.account__v = b.id
  where a.name__v in ('Denali HCO Survey', 'Denali HCP Survey')
),

ref_hco as (
  select *
  from com_edp_prd.cmpa_insights_internal_schema.reference_file_0109
),

ref_hcp as (
  select *
  from com_edp_prd.cmpa_insights_internal_schema.reference_file_0109
)

select
  s.npi,
  s.survey_name,
  case
    when s.survey_name = 'Denali HCO Survey' then rhco.territory
    when s.survey_name = 'Denali HCP Survey' then rhcp.territory
  end as territory_final,
  case
    when s.survey_name = 'Denali HCO Survey' then rhco.account_lead
    when s.survey_name = 'Denali HCP Survey' then rhcp.account_lead
  end as account_lead_final
from survey_spine s
left join ref_hco rhco
  on s.survey_name = 'Denali HCO Survey'
 and try_cast(s.npi as string) = try_cast(rhco.hco_npi as string)
left join ref_hcp rhcp
  on s.survey_name = 'Denali HCP Survey'
 and try_cast(s.npi as string) = try_cast(rhcp.hcp_npi as string)
;
```
