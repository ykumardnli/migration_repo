```sql
WITH t1 as (SELECT patient_id, service_date AS fill_date, ndc11 AS code, ingestion_date as date
      FROM com_edp_prd.com_raw.kom_medical_events
      WHERE ndc11 IN ('54092070001','540920700')

      UNION ALL
      SELECT patient_id, fill_date, ndc11 AS code, ingestion_date as date
      FROM com_edp_prd.com_raw.kom_pharmacy_events
      WHERE ndc11 IN ('54092070001','540920700')
        AND transaction_result = 'PAID'

      UNION ALL
      SELECT patient_id, service_date AS fill_date, procedure_code AS code, ingestion_date as date
      FROM com_edp_prd.com_raw.kom_medical_events
      WHERE procedure_code IN (
        '99601','99602','96365','96366','J1743','S9357','S9379',
        '38206','38230','38232','38240','38241','38242','38243','38250'
      ))

select * from t1
where patient_id in ('4VXJSWWG', '9YVY62P8', 'REQPDYJ6', '5Z4BWQ9S', 'RSD92JLE', 'YPHZSNH8', 'HNNWFT61', '7W4E0S3C', 'C6S52DKD', 'H6S5SSXY', 'BKVRTPS7', 'QLPDH514', '6K3MNJ8Z', '4QQ3B4V7') and fill_date between '2023-08-01' AND '2025-07-31'
```
