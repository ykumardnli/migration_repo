```sql
WITH tier2_clean AS (

SELECT
    hco_name,
    hco_zip,
    hco_state,
    UPPER(TRIM(hco_name)) AS tier2_name,
    UPPER(TRIM(hco_address)) AS tier2_address
FROM tier2_hcos

),

vod_hco_clean AS (

SELECT
    h.vid__v,
    h.npi_num__v,
    UPPER(TRIM(h.corporate_name__v)) AS vod_name,
    a.address_line_1__v,
    a.postal_code_cda__v,
    a.state_cda__v
FROM com_raw.vod_hco h
LEFT JOIN com_raw.vod_address a
  ON h.vid__v = a.entity_vid__v
 AND a.entity_type__v = 'HCO'
 AND a.record_state__v = 'VALID'
 AND a.address_status__v IN ('A','DS')

),

/* ------------------------------------------------
Match Tier2 HCO → VOD HCO
------------------------------------------------ */

tier2_vod_match AS (

SELECT
    t.hco_name AS tier2_hco_name,
    v.vid__v AS vod_vid,
    v.npi_num__v AS vod_npi,
    v.vod_name,
    v.postal_code_cda__v,
    v.state_cda__v,

    /* similarity using levenshtein */
    1 - (
        levenshtein(t.tier2_name, v.vod_name) /
        greatest(length(t.tier2_name), length(v.vod_name))
    ) AS name_similarity

FROM tier2_clean t
JOIN vod_hco_clean v
  ON t.hco_zip = v.postal_code_cda__v
 AND t.hco_state = v.state_cda__v

),

/* keep strong matches */

tier2_vod_filtered AS (

SELECT *
FROM tier2_vod_match
WHERE name_similarity > 0.80

),

/* ------------------------------------------------
Find parent VID
------------------------------------------------ */

parent_identification AS (

SELECT
    m.tier2_hco_name,
    h.vid__v AS matched_vid,

    CASE
        WHEN h.top_parent__v IS NOT NULL THEN h.top_parent__v
        WHEN h.immediate_parent__v IS NOT NULL THEN h.immediate_parent__v
        WHEN h.hospital_parent__v IS NOT NULL THEN h.hospital_parent__v
        ELSE h.vid__v
    END AS parent_vid

FROM tier2_vod_filtered m
LEFT JOIN com_raw.vod_hco h
  ON m.vod_vid = h.vid__v

),

/* ------------------------------------------------
Pull all children under the parent
------------------------------------------------ */

child_accounts AS (

SELECT
    p.tier2_hco_name,
    ph.npi_num__v AS parent_npi,
    ph.corporate_name__v AS parent_name,

    c.npi_num__v AS child_npi,
    c.corporate_name__v AS child_name

FROM parent_identification p

LEFT JOIN com_raw.vod_hco ph
  ON p.parent_vid = ph.vid__v

LEFT JOIN com_raw.vod_hco c
  ON c.top_parent__v = p.parent_vid

)

SELECT *
FROM child_accounts
ORDER BY parent_name, child_name;
```

```sql
WITH tier2_clean AS (

SELECT
    hco_npi,
    hco_name,
    hco_zip,
    hco_state,
    hco_address,
    UPPER(TRIM(hco_name)) clean_name,
    UPPER(TRIM(hco_address)) clean_address
FROM cmpa_insights_internal_schema.tier2_hcos
where hco_npi not in ('1023494473','1740215219','1073576740','1053635466','1447233788','1255461075','1811056898','1003968579','1184612764','1245520386','1417946021','1699720086','1871540237','1821017880','1447355771','1376577247','1295137404','1467525790','1407801640','1285641514','1447352836','1437119310','1396837951','1144313743','1487306692','1952359986','1780676650'
)

),

/* ---------------------------------------------------
VOD HCO + ADDRESS COMBINED
--------------------------------------------------- */

vod_full AS (

SELECT
    h.vid__v,
    h.corporate_name__v,
    h.npi_num__v,
    h.top_parent__v,
    UPPER(TRIM(h.corporate_name__v)) vod_name,
    a.address_line_1__v vod_address,
    UPPER(TRIM(a.address_line_1__v)) clean_vod_address,
    substr(a.postal_code_cda__v,1,5) vod_zip

FROM com_raw.vod_hco h

LEFT JOIN com_raw.vod_address a
ON h.vid__v = a.entity_vid__v
AND a.record_state__v='VALID'

),

/* ---------------------------------------------------
FUZZY MATCH + PARENT FILTER IN ONE STEP
--------------------------------------------------- */

matched_parents AS (

SELECT *
FROM (

SELECT
    t.hco_npi,
    t.hco_name,
    t.hco_address,
    t.hco_zip,

    /* fuzzy matched VOD account */
    v.vid__v matched_vid,
    v.npi_num__v matched_npi,
    v.corporate_name__v matched_vod_name,
    v.vod_address matched_vod_address,
    v.vod_zip matched_vod_zip,
    v.top_parent__v,

    /* parent account */
    p.vid__v parent_vid,
    p.corporate_name__v parent_name,
    p.npi_num__v parent_npi,

    /* similarity metrics */
    1 - (
        levenshtein(t.clean_name,v.vod_name) /
        greatest(length(t.clean_name),length(v.vod_name))
    ) AS name_similarity,

    1 - (
        levenshtein(t.clean_address,v.clean_vod_address) /
        greatest(length(t.clean_address),length(v.clean_vod_address))
    ) AS address_similarity,

    /* ranking to pick best match */
    ROW_NUMBER() OVER(
        PARTITION BY t.hco_name
        ORDER BY
        (
            1 - (levenshtein(t.clean_name,v.vod_name) /
            greatest(length(t.clean_name),length(v.vod_name)))
        ) DESC,

        (
            1 - (levenshtein(t.clean_address,v.clean_vod_address) /
            greatest(length(t.clean_address),length(v.clean_vod_address)))
        ) DESC
    ) AS rn

FROM tier2_clean t

JOIN vod_full v
ON t.hco_zip = v.vod_zip

LEFT JOIN com_raw.vod_hco p
ON v.top_parent__v = p.vid__v

INNER JOIN cmpa_insights_internal_schema.reference_file r
ON p.npi_num__v = r.hco_npi

) x

WHERE rn = 1

),
/* ---------------------------------------------------
CHILDREN FROM REFERENCE FILE
--------------------------------------------------- */

children AS (

SELECT
    h.top_parent__v parent_vid,
    h.vid__v child_vid,
    h.corporate_name__v child_name,
    h.npi_num__v child_npi

FROM com_raw.vod_hco h

INNER JOIN cmpa_insights_internal_schema.reference_file r
ON h.npi_num__v = r.hco_npi

),

/* ---------------------------------------------------
ADDRESS TABLE
--------------------------------------------------- */

addr AS (

SELECT
    entity_vid__v vid,
    address_line_1__v address,
    substr(postal_code_cda__v,1,5) zip

FROM com_raw.vod_address
WHERE record_state__v='VALID'
AND entity_type__v = 'HCO'
AND address_status__v IN ('A','DS')
AND address_verification_status__v NOT IN ('NS','U')

)

/* ---------------------------------------------------
FINAL OUTPUT
--------------------------------------------------- */

SELECT DISTINCT

m.hco_name tier2_hco_name,
m.hco_address tier2_hco_address,
m.hco_zip tier2_hco_zip,

/* fuzzy matched VOD account */
m.matched_vid,
m.matched_vod_name,
m.matched_npi,
m.matched_vod_address,
m.matched_vod_zip,
m.name_similarity,
m.address_similarity,

m.parent_vid,
m.parent_name,
m.parent_npi,

pa.address parent_address,
pa.zip parent_zip,

c.child_vid,
c.child_name,
c.child_npi,

ca.address child_address,
ca.zip child_zip,

CASE 
    WHEN t2.hco_npi = c.child_npi 
         OR 
         (
            UPPER(TRIM(t2.hco_name)) = UPPER(TRIM(c.child_name))
            and UPPER(TRIM(t2.clean_address)) = UPPER(TRIM(ca.address))
            and UPPER(TRIM(t2.hco_zip)) = UPPER(TRIM(ca.zip))
         )
    THEN 1
    ELSE 0
END AS child_account_in_tier2_flag

FROM matched_parents m

LEFT JOIN addr pa
ON m.parent_vid = pa.vid

INNER JOIN children c
ON m.parent_vid = c.parent_vid

LEFT JOIN addr ca
ON c.child_vid = ca.vid

LEFT JOIN tier2_clean t2
ON t2.hco_npi = c.child_npi
OR UPPER(TRIM(t2.hco_name)) = UPPER(TRIM(c.child_name));
```

```sql
/* ---------------------------------------------------
TIER2 CLEAN
Only HCOs WITHOUT NPI -> need fuzzy matching
--------------------------------------------------- */

WITH tier2_clean AS (

SELECT
    hco_npi,
    hco_name,
    hco_zip,
    hco_state,
    hco_address,

    UPPER(TRIM(hco_name)) clean_name,
    UPPER(TRIM(hco_address)) clean_address

FROM cmpa_insights_internal_schema.tier2_hcos
WHERE hco_npi IS NULL

),

/* ---------------------------------------------------
VOD HCO + ADDRESS
--------------------------------------------------- */

vod_full AS (

SELECT
    h.vid__v,
    h.corporate_name__v,
    h.npi_num__v,
    h.top_parent__v,

    UPPER(TRIM(h.corporate_name__v)) vod_name,

    a.address_line_1__v vod_address,
    UPPER(TRIM(a.address_line_1__v)) clean_vod_address,

    substr(a.postal_code_cda__v,1,5) vod_zip

FROM com_raw.vod_hco h

LEFT JOIN com_raw.vod_address a
ON h.vid__v = a.entity_vid__v
AND a.record_state__v = 'VALID'

),

/* ---------------------------------------------------
FUZZY MATCH
--------------------------------------------------- */

fuzzy_match AS (

SELECT

t.hco_name,
t.hco_address,
t.hco_zip,

v.vid__v matched_vid,
v.corporate_name__v matched_vod_name,
v.npi_num__v matched_npi,
v.top_parent__v,

v.vod_address,
v.vod_zip,

/* similarity scores */

1 - (
    levenshtein(t.clean_name,v.vod_name) /
    greatest(length(t.clean_name),length(v.vod_name))
) AS name_similarity,

1 - (
    levenshtein(t.clean_address,v.clean_vod_address) /
    greatest(length(t.clean_address),length(v.clean_vod_address))
) AS address_similarity,

/* combined score */

(
(
1 - (
    levenshtein(t.clean_name,v.vod_name) /
    greatest(length(t.clean_name),length(v.vod_name))
)
)
+
(
1 - (
    levenshtein(t.clean_address,v.clean_vod_address) /
    greatest(length(t.clean_address),length(v.clean_vod_address))
)
)
) / 2 AS similarity_score

FROM tier2_clean t

JOIN vod_full v
ON t.hco_zip = v.vod_zip

),

/* ---------------------------------------------------
TOP 3 MATCHES
--------------------------------------------------- */

top_matches AS (

SELECT *

FROM (

SELECT *,
ROW_NUMBER() OVER (
    PARTITION BY hco_name
    ORDER BY similarity_score DESC
) rn

FROM fuzzy_match

) x

WHERE rn <= 3

),

/* ---------------------------------------------------
PARENT HCO
--------------------------------------------------- */

parent_accounts AS (

SELECT
    t.*,

    p.vid__v parent_vid,
    p.corporate_name__v parent_name,
    p.npi_num__v parent_npi

FROM top_matches t

LEFT JOIN com_raw.vod_hco p
ON t.top_parent__v = p.vid__v

),

/* ---------------------------------------------------
CHILD ACCOUNTS
--------------------------------------------------- */

children AS (

SELECT
    h.top_parent__v parent_vid,
    h.vid__v child_vid,
    h.corporate_name__v child_name,
    h.npi_num__v child_npi

FROM com_raw.vod_hco h

),

/* ---------------------------------------------------
REFERENCE FILE FILTER
IMPORTANT STEP
--------------------------------------------------- */

reference_children AS (

SELECT
    c.*

FROM children c

INNER JOIN cmpa_insights_internal_schema.reference_file r
ON c.child_npi = r.hco_npi

),

/* ---------------------------------------------------
ADDRESS TABLE
--------------------------------------------------- */

addr AS (

SELECT
    entity_vid__v vid,
    address_line_1__v address,
    substr(postal_code_cda__v,1,5) zip

FROM com_raw.vod_address
WHERE record_state__v='VALID'

)

/* ---------------------------------------------------
FINAL OUTPUT
--------------------------------------------------- */

SELECT DISTINCT

p.hco_name tier2_hco_name,
p.hco_address tier2_hco_address,
p.hco_zip tier2_hco_zip,

p.rn fuzzy_rank,
p.similarity_score,
p.name_similarity,
p.address_similarity,

/* matched vod account */

p.matched_vid,
p.matched_vod_name,
p.matched_npi,
p.vod_address,
p.vod_zip,

/* parent account */

p.parent_vid,
p.parent_name,
p.parent_npi,

pa.address parent_address,
pa.zip parent_zip,

/* children filtered via reference file */

c.child_vid,
c.child_name,
c.child_npi,

ca.address child_address,
ca.zip child_zip

FROM parent_accounts p

LEFT JOIN addr pa
ON p.parent_vid = pa.vid

INNER JOIN reference_children c
ON p.parent_vid = c.parent_vid

LEFT JOIN addr ca
ON c.child_vid = ca.vid

ORDER BY
p.hco_name,
p.rn;
```

```sql
-- ======================================================================
-- TIER-2 (INPUT): Use your provided temp view or create it above.
-- ======================================================================

-- -- (Optional) If not already created, paste your CREATE OR REPLACE TEMP VIEW tier2_hcos ... here.

-- ======================================================================
-- PARAMETERS (optional) – uncomment to filter out very weak matches
-- ======================================================================
-- -- Example: minimum acceptable similarity thresholds
-- -- SET min_name_similarity = 0.70;
-- -- SET min_addr_similarity = 0.60;

-- ======================================================================
-- CORE PIPELINE
-- ======================================================================
CREATE OR REPLACE TEMP VIEW final_tier2_vod_top3 AS
WITH
/* ---------------------------------------------------
STATE MAP (if needed elsewhere)
--------------------------------------------------- */
state_map AS (
  SELECT 'ALABAMA' state_name,'AL' state_abbr UNION ALL
  SELECT 'ALASKA','AK' UNION ALL
  SELECT 'ARIZONA','AZ' UNION ALL
  SELECT 'ARKANSAS','AR' UNION ALL
  SELECT 'CALIFORNIA','CA' UNION ALL
  SELECT 'COLORADO','CO' UNION ALL
  SELECT 'CONNECTICUT','CT' UNION ALL
  SELECT 'DELAWARE','DE' UNION ALL
  SELECT 'FLORIDA','FL' UNION ALL
  SELECT 'GEORGIA','GA' UNION ALL
  SELECT 'HAWAII','HI' UNION ALL
  SELECT 'IDAHO','ID' UNION ALL
  SELECT 'ILLINOIS','IL' UNION ALL
  SELECT 'INDIANA','IN' UNION ALL
  SELECT 'IOWA','IA' UNION ALL
  SELECT 'KANSAS','KS' UNION ALL
  SELECT 'KENTUCKY','KY' UNION ALL
  SELECT 'LOUISIANA','LA' UNION ALL
  SELECT 'MAINE','ME' UNION ALL
  SELECT 'MARYLAND','MD' UNION ALL
  SELECT 'MASSACHUSETTS','MA' UNION ALL
  SELECT 'MICHIGAN','MI' UNION ALL
  SELECT 'MINNESOTA','MN' UNION ALL
  SELECT 'MISSISSIPPI','MS' UNION ALL
  SELECT 'MISSOURI','MO' UNION ALL
  SELECT 'MONTANA','MT' UNION ALL
  SELECT 'NEBRASKA','NE' UNION ALL
  SELECT 'NEVADA','NV' UNION ALL
  SELECT 'NEW HAMPSHIRE','NH' UNION ALL
  SELECT 'NEW JERSEY','NJ' UNION ALL
  SELECT 'NEW MEXICO','NM' UNION ALL
  SELECT 'NEW YORK','NY' UNION ALL
  SELECT 'NORTH CAROLINA','NC' UNION ALL
  SELECT 'NORTH DAKOTA','ND' UNION ALL
  SELECT 'OHIO','OH' UNION ALL
  SELECT 'OKLAHOMA','OK' UNION ALL
  SELECT 'OREGON','OR' UNION ALL
  SELECT 'PENNSYLVANIA','PA' UNION ALL
  SELECT 'RHODE ISLAND','RI' UNION ALL
  SELECT 'SOUTH CAROLINA','SC' UNION ALL
  SELECT 'SOUTH DAKOTA','SD' UNION ALL
  SELECT 'TENNESSEE','TN' UNION ALL
  SELECT 'TEXAS','TX' UNION ALL
  SELECT 'UTAH','UT' UNION ALL
  SELECT 'VERMONT','VT' UNION ALL
  SELECT 'VIRGINIA','VA' UNION ALL
  SELECT 'WASHINGTON','WA' UNION ALL
  SELECT 'WEST VIRGINIA','WV' UNION ALL
  SELECT 'WISCONSIN','WI' UNION ALL
  SELECT 'WYOMING','WY'
),

/* ---------------------------------------------------
TIER-2 CLEAN: Only HCOs with NULL NPI (to be fuzzy matched)
--------------------------------------------------- */
tier2_clean AS (
  SELECT
      hco_npi,
      hco_name,
      hco_zip,
      hco_address,
      UPPER(TRIM(COALESCE(hco_name,'')))     AS clean_name,
      UPPER(TRIM(COALESCE(hco_address,'')))  AS clean_address
  FROM tier2_hcos
  WHERE hco_npi IS NULL
),

/* ---------------------------------------------------
VOD HCO + ADDRESS with quality filters (builds on "proper" query)
--------------------------------------------------- */
vod_full AS (
  SELECT
      h.vid__v,
      h.corporate_name__v,
      h.npi_num__v,
      h.top_parent__v,
      UPPER(TRIM(COALESCE(h.corporate_name__v, ''))) AS vod_name,
      a.address_line_1__v                              AS vod_address,
      UPPER(TRIM(COALESCE(a.address_line_1__v, '')))  AS clean_vod_address,
      SUBSTR(a.postal_code_cda__v, 1, 5)              AS vod_zip
  FROM com_raw.vod_hco h
  LEFT JOIN com_raw.vod_address a
    ON h.vid__v = a.entity_vid__v
   AND a.record_state__v = 'VALID'
   AND a.entity_type__v = 'HCO'
   AND a.address_status__v IN ('A','DS')
   AND a.address_verification_status__v NOT IN ('NS','U')
),

/* ---------------------------------------------------
FUZZY MATCH SCORING (zip pre-filter), no hard threshold
--------------------------------------------------- */
tier2_vod_scored AS (
  SELECT
      t.*,
      v.vid__v                 AS matched_vid,
      v.npi_num__v             AS matched_npi,
      v.corporate_name__v      AS matched_vod_name,
      v.vod_address            AS matched_vod_address,
      v.vod_zip                AS matched_vod_zip,
      v.top_parent__v,

      -- Name similarity: normalized Levenshtein; safe for blanks
      CASE
        WHEN GREATEST(LENGTH(t.clean_name), LENGTH(v.vod_name)) = 0 THEN 0.0
        ELSE 1.0 - (LEVENSHTEIN(t.clean_name, v.vod_name) /
                    CAST(GREATEST(LENGTH(t.clean_name), LENGTH(v.vod_name)) AS DOUBLE))
      END AS name_similarity,

      -- Address similarity: normalized Levenshtein; safe for blanks
      CASE
        WHEN GREATEST(LENGTH(t.clean_address), LENGTH(v.clean_vod_address)) = 0 THEN 0.0
        ELSE 1.0 - (LEVENSHTEIN(t.clean_address, v.clean_vod_address) /
                    CAST(GREATEST(LENGTH(t.clean_address), LENGTH(v.clean_vod_address)) AS DOUBLE))
      END AS address_similarity
  FROM tier2_clean t
  JOIN vod_full v
    ON t.hco_zip = v.vod_zip
  -- Optional: uncomment to enforce minimum match quality
  -- WHERE
  --   (CASE
  --     WHEN GREATEST(LENGTH(t.clean_name), LENGTH(v.vod_name)) = 0 THEN 0.0
  --     ELSE 1.0 - (LEVENSHTEIN(t.clean_name, v.vod_name) /
  --                 CAST(GREATEST(LENGTH(t.clean_name), LENGTH(v.vod_name)) AS DOUBLE))
  --   END) >= ${min_name_similarity}
  -- AND
  --   (CASE
  --     WHEN GREATEST(LENGTH(t.clean_address), LENGTH(v.clean_vod_address)) = 0 THEN 0.0
  --     ELSE 1.0 - (LEVENSHTEIN(t.clean_address, v.clean_vod_address) /
  --                 CAST(GREATEST(LENGTH(t.clean_address), LENGTH(v.clean_vod_address)) AS DOUBLE))
  --   END) >= ${min_addr_similarity}
),

/* ---------------------------------------------------
TOP 3 MATCHES per Tier-2 HCO (by name DESC, then address DESC)
--------------------------------------------------- */
top3_matches AS (
  SELECT
      s.*,
      ROW_NUMBER() OVER (
        PARTITION BY
          s.hco_name, s.hco_address, s.hco_zip
        ORDER BY
          s.name_similarity DESC,
          s.address_similarity DESC
      ) AS rn
  FROM tier2_vod_scored s
),
best3 AS (
  SELECT *
  FROM top3_matches
  WHERE rn <= 3
),

/* ---------------------------------------------------
ATTACH PARENT (and enforce parent is in reference file, like your proper query)
--------------------------------------------------- */
ranked_with_parent AS (
  SELECT
      b.*,
      p.vid__v            AS parent_vid,
      p.corporate_name__v AS parent_name,
      p.npi_num__v        AS parent_npi
  FROM best3 b
  LEFT JOIN com_raw.vod_hco p
    ON b.top_parent__v = p.vid__v
  INNER JOIN cmpa_insights_internal_schema.reference_file r_parent
    ON p.npi_num__v = r_parent.hco_npi
),

/* ---------------------------------------------------
PARENT ADDRESS (quality filters)
--------------------------------------------------- */
parent_addr AS (
  SELECT
      a.entity_vid__v                     AS parent_vid,
      a.address_line_1__v                 AS parent_address,
      SUBSTR(a.postal_code_cda__v, 1, 5)  AS parent_zip
  FROM com_raw.vod_address a
  WHERE a.record_state__v = 'VALID'
    AND a.entity_type__v = 'HCO'
    AND a.address_status__v IN ('A','DS')
    AND a.address_verification_status__v NOT IN ('NS','U')
),

/* ---------------------------------------------------
CHILDREN of Parent (KEEP ONLY those present in reference_file)
--------------------------------------------------- */
children_ref AS (
  SELECT
      h.top_parent__v          AS parent_vid,
      h.vid__v                 AS child_vid,
      h.corporate_name__v      AS child_name,
      h.npi_num__v             AS child_npi,
      a.address_line_1__v      AS child_address,
      SUBSTR(a.postal_code_cda__v, 1, 5) AS child_zip
  FROM com_raw.vod_hco h
  LEFT JOIN com_raw.vod_address a
    ON h.vid__v = a.entity_vid__v
   AND a.record_state__v = 'VALID'
   AND a.entity_type__v = 'HCO'
   AND a.address_status__v IN ('A','DS')
   AND a.address_verification_status__v NOT IN ('NS','U')
  INNER JOIN cmpa_insights_internal_schema.reference_file r_child
    ON h.npi_num__v = r_child.hco_npi
)

/* ---------------------------------------------------
FINAL OUTPUT
--------------------------------------------------- */
SELECT
    m.hco_name              AS tier2_hco_name,
    m.hco_address           AS tier2_hco_address,
    m.hco_zip               AS tier2_hco_zip,

    -- Fuzzy-matched VOD account (Top 3 per Tier-2)
    m.matched_vid,
    m.matched_vod_name,
    m.matched_npi,
    m.matched_vod_address,
    m.matched_vod_zip,
    ROUND(m.name_similarity,    4) AS name_similarity,
    ROUND(m.address_similarity, 4) AS address_similarity,
    m.rn                        AS match_rank,  -- 1..3

    -- Parent details (from VOD)
    m.parent_vid,
    m.parent_name,
    m.parent_npi,
    pa.parent_address,
    pa.parent_zip,

    -- Child accounts (restricted to reference file)
    c.child_vid,
    c.child_name,
    c.child_npi,
    c.child_address,
    c.child_zip,

    -- Flag: does this Tier-2 HCO equal this child (name+address+zip)?
    CASE
      WHEN
        UPPER(TRIM(COALESCE(m.hco_name,    ''))) = UPPER(TRIM(COALESCE(c.child_name,    '')))
        AND UPPER(TRIM(COALESCE(m.hco_address, ''))) = UPPER(TRIM(COALESCE(c.child_address, '')))
        AND COALESCE(m.hco_zip, '') = COALESCE(c.child_zip, '')
      THEN 1
      ELSE 0
    END AS child_account_in_tier2_flag

FROM ranked_with_parent m
LEFT JOIN parent_addr pa
  ON m.parent_vid = pa.parent_vid
INNER JOIN children_ref c
  ON m.parent_vid = c.parent_vid
;
```
