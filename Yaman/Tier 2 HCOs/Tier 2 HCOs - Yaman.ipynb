```sql
WITH vod_presence AS (
  SELECT
    a.*,
    CASE WHEN b.npi_num__v IS NOT NULL THEN 1 ELSE 0 END AS vod_presence
  FROM tier2_hcos_full AS a
  LEFT JOIN com_raw.vod_hco AS b
    ON a.hco_npi = b.npi_num__v
),

parent_vid_identification AS (
  SELECT
    a.vid__v       AS hco_vid,
    a.npi_num__v   AS hco_npi,
    a.top_parent__v AS parent_vid
  FROM com_raw.vod_hco AS a
  WHERE a.npi_num__v IN (
      SELECT DISTINCT hco_npi
      FROM vod_presence
      WHERE vod_presence = 1
    )
    AND a.top_parent__v IS NOT NULL
),

child_hcos_scraping AS (
  SELECT
    a.top_parent__v      AS parent_vid,
    a.vid__v             AS child_vid,
    a.npi_num__v         AS child_npi,
    a.corporate_name__v  AS child_name,
    CASE WHEN b.hco_npi IS NOT NULL THEN 1 ELSE 0 END AS primary_flag
  FROM com_raw.vod_hco AS a
  LEFT JOIN tier2_hcos_full AS b
    ON a.npi_num__v = b.hco_npi
  WHERE a.top_parent__v IN (
    SELECT DISTINCT parent_vid
    FROM parent_vid_identification
  )
),

child_hcos_filtering_on_basis_of_hco_universe AS (
  SELECT *
  FROM child_hcos_scraping
  WHERE child_npi IN (
    SELECT DISTINCT hco_npi
    FROM cmpa_insights_internal_schema.reference_file
    WHERE hco_npi != '-'
      AND hco_npi IS NOT NULL
  )
),

/* ---------------------------------------------------------------------------
   Latest VALID HCO address per entity_vid__v (pattern copied from your 2nd query)
   --------------------------------------------------------------------------- */
hco_address_v1 AS (
  SELECT
    b.entity_vid__v                    AS hco_vid,
    b.address_line_1__v                AS hco_address,
    b.postal_code_cda__v               AS hco_postal_code,
    b.modified_date__v,
    ROW_NUMBER() OVER (
      PARTITION BY b.entity_vid__v
      ORDER BY b.modified_date__v DESC
    ) AS rn
  FROM com_raw.vod_address b
  WHERE b.entity_type__v = 'HCO'
    AND b.record_state__v = 'VALID'
    AND b.address_status__v IN ('A','DS')
    AND b.address_verification_status__v NOT IN ('NS','U')
),

hco_address_latest AS (
  SELECT
    hco_vid,
    hco_address,
    hco_postal_code
  FROM hco_address_v1
  WHERE rn = 1
),

parent_details AS (
  SELECT DISTINCT
    a.parent_vid,
    b.npi_num__v          AS parent_npi,
    b.corporate_name__v   AS parent_name,

    a.child_vid,
    a.child_npi,
    a.child_name,
    a.primary_flag,

    -- parent address/zip
    pa.hco_address        AS parent_address,
    pa.hco_postal_code    AS parent_zip,

    -- child address/zip
    ca.hco_address        AS child_address,
    ca.hco_postal_code    AS child_zip

  FROM child_hcos_filtering_on_basis_of_hco_universe AS a
  LEFT JOIN com_raw.vod_hco AS b
    ON a.parent_vid = b.vid__v

  LEFT JOIN hco_address_latest pa
    ON pa.hco_vid = a.parent_vid

  LEFT JOIN hco_address_latest ca
    ON ca.hco_vid = a.child_vid
)

SELECT *
FROM parent_details;
```

```sql
-- ==========================================
-- Top-3 VOD matches per Tier2 HCO (Databricks)
-- Uses latest valid VOD address (rn=1)
-- ==========================================

with tier2_hcos as (
  select *
  from (select a.*,
  case when b.npi_num__v is not null then 1 else 0 end as vod_presence
  from tier2_hcos_full as a
  left join com_raw.vod_hco as b on a.hco_npi = b.npi_num__v)
  where vod_presence = 0
),

-- ==========================================
-- Top-3 VOD matches per Tier2 HCO (Databricks)
-- Uses latest valid VOD address (rn=1)
-- ==========================================

tier2_clean AS (
  SELECT
    hco_npi,
    hco_name,
    hco_zip,
    hco_state,
    hco_address,

    upper(regexp_replace(trim(coalesce(hco_name,'')),    '[^A-Z0-9 ]', '')) AS clean_name,
    upper(regexp_replace(trim(coalesce(hco_address,'')), '[^A-Z0-9 ]', '')) AS clean_address,
    substring(coalesce(hco_zip,''), 1, 5) AS clean_zip
  FROM tier2_hcos
),

vod_hco_clean AS (
  SELECT
    vid__v,
    corporate_name__v,
    npi_num__v,
    top_parent__v,
    upper(regexp_replace(trim(coalesce(corporate_name__v,'')), '[^A-Z0-9 ]', '')) AS vod_name
  FROM com_raw.vod_hco
),

-- Latest VALID address per HCO (pattern aligned with your example)
vod_addr_latest AS (
  SELECT
    b.entity_vid__v AS vid__v,
    b.address_line_1__v AS vod_address,
    upper(regexp_replace(trim(coalesce(b.address_line_1__v,'')), '[^A-Z0-9 ]', '')) AS clean_vod_address,
    substring(coalesce(b.postal_code_cda__v,''), 1, 5) AS vod_zip,
    b.modified_date__v,
    row_number() OVER (
      PARTITION BY b.entity_vid__v
      ORDER BY b.modified_date__v DESC
    ) AS rn
  FROM com_raw.vod_address b
  WHERE b.entity_type__v = 'HCO'
    AND b.record_state__v = 'VALID'
    AND b.address_status__v IN ('A','DS')
    AND b.address_verification_status__v NOT IN ('NS','U')
),

vod_full AS (
  SELECT
    h.vid__v,
    h.corporate_name__v,
    h.npi_num__v,
    h.top_parent__v,
    h.vod_name,
    a.vod_address,
    a.clean_vod_address,
    a.vod_zip
  FROM vod_hco_clean h
  LEFT JOIN vod_addr_latest a
    ON h.vid__v = a.vid__v
   AND a.rn = 1
),

-- Candidate generation:
-- If Tier2 zip exists -> match candidates by zip
-- Else -> match candidates by name prefix to avoid huge explosion
candidates AS (
  SELECT
    t.*,
    v.vid__v              AS matched_vid,
    v.corporate_name__v   AS matched_hco_name,
    v.npi_num__v          AS matched_npi,
    v.vod_address         AS matched_address,
    v.vod_zip             AS matched_zip,
    v.vod_name            AS matched_clean_name,
    v.clean_vod_address   AS matched_clean_address
  FROM tier2_clean t
  LEFT JOIN vod_full v
    ON (
         (t.clean_zip <> '' AND t.clean_zip = v.vod_zip)
      OR (t.clean_zip = ''  AND substring(t.clean_name, 1, 6) = substring(v.vod_name, 1, 6))
       )
),

scored AS (
  SELECT
    c.*,

    CASE
      WHEN c.clean_name = '' OR c.matched_clean_name = '' THEN NULL
      ELSE 1 - (
        levenshtein(c.clean_name, c.matched_clean_name) /
        NULLIF(greatest(length(c.clean_name), length(c.matched_clean_name)), 0)
      )
    END AS name_similarity,

    CASE
      WHEN c.clean_address = '' OR c.matched_clean_address = '' THEN NULL
      ELSE 1 - (
        levenshtein(c.clean_address, c.matched_clean_address) /
        NULLIF(greatest(length(c.clean_address), length(c.matched_clean_address)), 0)
      )
    END AS address_similarity

  FROM candidates c
),

ranked AS (
  SELECT
    s.*,
    row_number() OVER (
      PARTITION BY
        -- stable Tier2 key
        s.hco_name, s.hco_zip, s.hco_address, s.hco_npi
      ORDER BY
        coalesce(s.name_similarity, 0.0) DESC,
        coalesce(s.address_similarity, 0.0) DESC,
        CASE WHEN s.hco_npi IS NOT NULL AND s.hco_npi = s.matched_npi THEN 1 ELSE 0 END DESC
    ) AS match_rank
  FROM scored s
  WHERE s.matched_vid IS NOT NULL
    AND coalesce(s.name_similarity, 0.0) >= 0.80
    AND (
          s.clean_address = '' OR s.matched_clean_address = ''
          OR coalesce(s.address_similarity, 0.0) >= 0.80
        )
)

SELECT
  hco_npi        AS tier2_hco_npi,
  hco_name       AS tier2_hco_name,
  hco_address    AS tier2_hco_address,
  hco_zip        AS tier2_hco_zip,
  hco_state      AS tier2_hco_state,

  matched_vid,
  matched_hco_name,
  matched_npi,
  matched_address,
  matched_zip,

  name_similarity,
  address_similarity,
  match_rank
FROM ranked
WHERE match_rank <= 3
ORDER BY tier2_hco_name, match_rank;
```
