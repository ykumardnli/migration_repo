### Step 2.1: Creating `elaprase_closed_claims`

Temporary view of **Elaprase treatment claims** across medical and pharmacy sources.

A temp view that collects **Elaprase claims** from medical and pharmacy data.

**Includes:**

* medical claims with Elaprase NDCs
* pharmacy claims with Elaprase NDCs that are paid/finalized
* medical claims with J-code `J1743`

**Filters:**

* only claims between `${start_date}` and `${end_date}`

**Why it exists:**

* to create one clean table of Elaprase treatment claims for later cohort-building steps




```sql
-- Table: Elaprase Closed Claims

CREATE OR REPLACE TEMPORARY VIEW elaprase_closed_claims AS
SELECT *
FROM (
    -- =========================
    -- MEDICAL (NDC)
    -- =========================
    SELECT DISTINCT
        PATIENT_ID AS PATIENT_ID,
        COALESCE(RENDERING_NPI, REFERRING_NPI) AS NPI,
        NDC11 AS CODE,
        MEDICAL_EVENT_ID AS EVENT_ID,
        SERVICE_DATE AS FILL_DATE,
        PLACE_OF_SERVICE,
        KH_PLAN_ID AS KH_PLAN,
        'MEDICAL_EVENTS' AS TABLE_NAME
    FROM com_edp_prd.com_raw.kom_medical_events
    WHERE NDC11 IN ('54092070001','540920700')

    UNION

    -- =========================
    -- PHARMACY (NDC)
    -- =========================
    SELECT DISTINCT
        PATIENT_ID AS PATIENT_ID,
        PRESCRIBER_NPI AS NPI,
        NDC11 AS CODE,
        PHARMACY_EVENT_ID AS EVENT_ID,
        FILL_DATE,
        NULL AS PLACE_OF_SERVICE,
        COALESCE(PRIMARY_KH_PLAN_ID, SECONDARY_KH_PLAN_ID) AS KH_PLAN,
        'PHARMACY_EVENTS' AS TABLE_NAME
    FROM com_edp_prd.com_raw.kom_pharmacy_events
    WHERE NDC11 IN ('54092070001','540920700')
    and TRANSACTION_STATUS IN ('Final', 'Standalone') AND TRANSACTION_RESULT = 'Paid'

    UNION

    -- =========================
    -- MEDICAL (J-CODE)
    -- =========================
    SELECT DISTINCT
        PATIENT_ID AS PATIENT_ID,
        RENDERING_NPI AS NPI,
        PROCEDURE_CODE AS CODE,
        MEDICAL_EVENT_ID AS EVENT_ID,
        SERVICE_DATE AS FILL_DATE,
        PLACE_OF_SERVICE,
        KH_PLAN_ID AS KH_PLAN,
        'MEDICAL_EVENTS' AS TABLE_NAME
    FROM com_edp_prd.com_raw.kom_medical_events
    WHERE PROCEDURE_CODE IN ('J1743')
)
WHERE FILL_DATE BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}');
```

## Step 2.2: Creating `diagnosis_closed_claims`

A temp view that collects **diagnosis claims for E761** from medical and pharmacy data.

**Includes:**

* medical claims where `DIAGNOSIS_CODES` contains `E761`
* pharmacy claims where `DIAGNOSIS_CODE = 'E761'` and the claim is paid/finalized

**Filters:**

* only claims between `${start_date}` and `${end_date}`

**Why it exists:**

* to create one clean table of E761 diagnosis claims for later cohort-building steps


```sql
-- Table: Diagnosis Closed Claims
CREATE OR REPLACE TEMPORARY VIEW diagnosis_closed_claims AS 
SELECT * FROM (
    SELECT DISTINCT 
      PATIENT_ID,
      COALESCE(RENDERING_NPI, REFERRING_NPI) AS NPI,
      SERVICE_DATE AS FILL_DATE,
      MEDICAL_EVENT_ID AS EVENT_ID,
      DIAGNOSIS_CODES,
      KH_PLAN_ID AS KH_PLAN,
      Place_of_service
    FROM com_edp_prd.com_raw.kom_medical_events
    WHERE DIAGNOSIS_CODES LIKE '%E761%'

    UNION

    -- Pharmacy Events
    SELECT DISTINCT 
      PATIENT_ID,
      PRESCRIBER_NPI AS NPI,
      FILL_DATE,
      PHARMACY_EVENT_ID AS EVENT_ID,
      DIAGNOSIS_CODE AS DIAGNOSIS_CODES,
      COALESCE(PRIMARY_KH_PLAN_ID, SECONDARY_KH_PLAN_ID) AS KH_PLAN,
      NULL as Place_of_service
    FROM com_edp_prd.com_raw.kom_pharmacy_events
    WHERE DIAGNOSIS_CODE = 'E761'
    and TRANSACTION_STATUS IN ('Final', 'Standalone') AND TRANSACTION_RESULT = ('Paid')
  ) AS combined
  WHERE FILL_DATE BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}');
```

### Step 3.2: Creating `continuous_enrollment_patients_pm6m`

A temp view that finds patients with **continuous insurance enrollment from 6 months before to 6 months after** first Elaprase treatment.

**Logic:**

* uses first Elaprase treatment date as the index date
* builds a required coverage window: **index date - 6 months** to **index date + 6 months**
* merges overlapping or back-to-back enrollment periods
* checks whether one merged enrollment block fully covers that whole window

**What it outputs:**

* `patient_id` values for patients who meet continuous enrollment in the ±6 month window

**Why it exists:**

* to keep only patients with complete enrollment coverage around treatment for cohort eligibility


```sql
-- Table: continuous_enrollment_patients_pm6m 

-- =============================================================================
-- Continuous enrollment check for ONLY: index_date - 6 months  to  index_date + 6 months
-- Business rules :
--   • merge overlapping/adjacent enrollment rows (<= 1 day gap allowed)
--   • pick the merged block that covers index_date
--   • flag = 1 if that block fully covers [index-6mo, index+6mo]
-- Output: distinct patient_id with continuous coverage in +/- 6 months window
-- =============================================================================

CREATE OR REPLACE TEMPORARY VIEW continuous_enrollment_patients_pm6m AS
WITH idx AS (
  SELECT
    patient_id,
    first_elaprase_treatment_date AS index_date,

    -- window bounds: +/- 6 months
    add_months(first_elaprase_treatment_date, -6) AS pre_6m_start,
    add_months(first_elaprase_treatment_date,  6) AS post_6m_end

  FROM first_elaprase_treatment
),

-- only enrollment rows that could affect the +/- 6 month window
enr AS (
  SELECT
    e.patient_id,
    e.start_date,
    e.end_date
  FROM com_raw.kom_patient_enrollment e
  JOIN idx i
    ON e.patient_id = i.patient_id
  WHERE e.end_date  >= i.pre_6m_start
    AND e.start_date <= i.post_6m_end
),

-- detect gaps (gaps & islands) with 1-day adjacency allowed (same rule as above)
ordered AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    CASE
      WHEN lag(end_date) OVER (
             PARTITION BY patient_id ORDER BY start_date, end_date
           ) IS NULL
      THEN 1
      WHEN start_date <= date_add(
             lag(end_date) OVER (PARTITION BY patient_id ORDER BY start_date, end_date),
             1
           )
      THEN 0
      ELSE 1
    END AS new_block
  FROM enr
),

blocks AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    sum(new_block) OVER (
      PARTITION BY patient_id ORDER BY start_date, end_date
      ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS block_id
  FROM ordered
),

merged AS (
  SELECT
    patient_id,
    block_id,
    min(start_date) AS block_start,
    max(end_date)   AS block_end
  FROM blocks
  GROUP BY patient_id, block_id
),

-- get the block that covers index_date
covering AS (
  SELECT
    i.patient_id,
    i.index_date,
    i.pre_6m_start,
    i.post_6m_end,
    m.block_start,
    m.block_end
  FROM idx i
  LEFT JOIN merged m
    ON i.patient_id = m.patient_id
   AND i.index_date BETWEEN m.block_start AND m.block_end
),

flags AS (
  SELECT
    patient_id,
    index_date,
    CASE
      WHEN block_start <= pre_6m_start
       AND block_end   >= post_6m_end
      THEN 1 ELSE 0
    END AS cont_pm_6m_flag
  FROM covering
)

SELECT DISTINCT patient_id
FROM flags
WHERE cont_pm_6m_flag = 1;
```

### Step 4.1: Pulling all Comorbidity Claims

```sql
CREATE OR REPLACE TABLE cmpa_insights_internal_schema.comorbidity_claims AS
WITH code_map AS (
  SELECT DISTINCT
    new_bucket,
    subcategory,
    UPPER(TRIM(code)) AS code,
    code_type,
    CASE
      WHEN code_type IN ('ICD-9-CM Diagnosis', 'ICD-10-CM Diagnosis') THEN 'DX'
      WHEN code_type IN ('CPT', 'HCPCS Level II') THEN 'PROC'
      ELSE 'OTHER'
    END AS match_domain
  FROM codes_map_v1
),

dx_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'DX'
),

proc_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'PROC'
),

medical_base AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
),

medical_dx_exploded AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type,
    REGEXP_REPLACE(UPPER(TRIM(dx_token)), '\\.', '') AS dx_code
  FROM medical_base
  LATERAL VIEW OUTER explode(split(coalesce(diagnosis_codes,''), '\\|')) dx AS dx_token
),

medical_dx AS (
  SELECT
    me.patient_id,
    me.medical_event_id AS claim_id,
    me.service_date      AS fill_date,
    'MEDICAL'            AS source,
    me.diagnosis_codes   AS diagnosis_codes,
    me.procedure_code    AS procedure_code,
    me.procedure_code_type AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM medical_dx_exploded me
  JOIN dx_map dm
    ON me.dx_code = REGEXP_REPLACE(dm.code, '\\.', '')
  WHERE me.dx_code IS NOT NULL AND me.dx_code <> ''
),

medical_proc AS (
  SELECT
    mb.patient_id,
    mb.medical_event_id AS claim_id,
    mb.service_date      AS fill_date,
    'MEDICAL'            AS source,
    mb.diagnosis_codes   AS diagnosis_codes,
    mb.procedure_code    AS procedure_code,
    mb.procedure_code_type AS procedure_code_type,
    'PROC'               AS types,
    pm.code              AS matched_code,
    pm.new_bucket        AS matched_new_bucket,
    pm.subcategory       AS matched_subcategory
  FROM medical_base mb
  JOIN proc_map pm
    ON UPPER(TRIM(mb.procedure_code)) = pm.code
),

pharmacy_base AS (
  SELECT
    patient_id,
    pharmacy_event_id,
    fill_date,
    diagnosis_code
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
    AND transaction_status IN ('Final', 'Standalone')
    AND transaction_result = 'Paid'
),

pharmacy_dx AS (
  SELECT
    pb.patient_id,
    pb.pharmacy_event_id AS claim_id,
    pb.fill_date         AS fill_date,
    'PHARMACY'           AS source,
    pb.diagnosis_code    AS diagnosis_codes,
    CAST(NULL AS STRING) AS procedure_code,
    CAST(NULL AS STRING) AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM pharmacy_base pb
  JOIN dx_map dm
    ON REGEXP_REPLACE(UPPER(TRIM(pb.diagnosis_code)), '\\.', '') = REGEXP_REPLACE(dm.code, '\\.', '')
),

final_union AS (
  SELECT * FROM medical_dx
  UNION ALL
  SELECT * FROM medical_proc
  UNION ALL
  SELECT * FROM pharmacy_dx
),

qc_check AS (
  SELECT
    t.*,
    CASE
      WHEN t.matched_code IS NULL OR TRIM(t.matched_code) = '' THEN 0
      WHEN LOWER(t.diagnosis_codes) LIKE '%|' || LOWER(t.matched_code) || '|%'
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) = LOWER(t.matched_code)
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) LIKE '%' || LOWER(t.matched_code) || '%'
      THEN 1
      ELSE 0
    END AS matched_code_present
  FROM final_union t
),

-- ✅ 2-claim rule: patient must have >= 2 DISTINCT fill_date for the SAME matched_code XXXXX it should be for the SAME matched_subcategory------ updated code here
-- ✅ NEW (subcategory-level)
qualifying_patient_subcategory AS (
  SELECT
    patient_id,
    matched_new_bucket,
    matched_subcategory
  FROM qc_check
  WHERE matched_code_present = 1
  GROUP BY
    patient_id,
    matched_new_bucket,
    matched_subcategory
  HAVING COUNT(DISTINCT fill_date) >= 2
),

filtered_claims AS (
  SELECT q.*
  FROM qc_check q
  JOIN qualifying_patient_subcategory k
    ON q.patient_id = k.patient_id
   AND q.matched_new_bucket = k.matched_new_bucket
   AND q.matched_subcategory = k.matched_subcategory
  WHERE q.matched_code_present = 1
),

patient_age AS (
  SELECT
    a.*,
    (year(current_date) - year(b.patient_yob)) AS age
  FROM filtered_claims a
  LEFT JOIN (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
      SELECT
        PATIENT_ID AS patient_id,
        PATIENT_YOB AS patient_yob,
        row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
      FROM com_raw.kom_patient_demographics
    )
    WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
)

SELECT * FROM patient_age;
```

```sql
SELECT
    age_bucket,
    COUNT(DISTINCT patient_id) AS patient_count,
    ROUND(COUNT(DISTINCT patient_id) * 100.0 / SUM(COUNT(DISTINCT patient_id)) OVER (), 2) AS pct_of_total
FROM (
    SELECT
        patient_id,
        age,
        CASE
            WHEN age BETWEEN 0 AND 5   THEN '0-5'
            WHEN age BETWEEN 6 AND 11  THEN '6-11'
            WHEN age BETWEEN 12 AND 16 THEN '12-16'
            WHEN age = 17              THEN '17'
            WHEN age >= 18             THEN '18+'
            ELSE 'Unknown'
        END AS age_bucket
    FROM cmpa_insights_internal_schema.comorbidity_claims
) aged
GROUP BY age_bucket
ORDER BY
    CASE age_bucket
        WHEN '0-5'    THEN 1
        WHEN '6-11'   THEN 2
        WHEN '12-16'  THEN 3
        WHEN '17'     THEN 4
        WHEN '18+'    THEN 5
        ELSE 6
    END;
```

```sql
SELECT
    age_bucket,
    COUNT(DISTINCT patient_id) AS patient_count,
    ROUND(COUNT(DISTINCT patient_id) * 100.0 / SUM(COUNT(DISTINCT patient_id)) OVER (), 2) AS pct_of_total
FROM (
    SELECT
        c.patient_id,
        c.age,
        CASE
            WHEN c.age BETWEEN 0 AND 5   THEN '0-5'
            WHEN c.age BETWEEN 6 AND 11  THEN '6-11'
            WHEN c.age BETWEEN 12 AND 16 THEN '12-16'
            WHEN c.age = 17              THEN '17'
            WHEN c.age >= 18             THEN '18+'
            ELSE 'Unknown'
        END AS age_bucket
    FROM cmpa_insights_internal_schema.comorbidity_claims c
    INNER JOIN eligible_patient_cohort e          -- 👈 only keeps patients present in cohort
        ON c.patient_id = e.patient_id
) aged
GROUP BY age_bucket
ORDER BY
    CASE age_bucket
        WHEN '0-5'    THEN 1
        WHEN '6-11'   THEN 2
        WHEN '12-16'  THEN 3
        WHEN '17'     THEN 4
        WHEN '18+'    THEN 5
        ELSE 6
    END;
```

```sql
-- =============================================================================
-- Combined table with:
--  • Eligible cohort denominators
--  • Whole number percentages
--  • "Overall" instead of NULL for bucket totals
-- =============================================================================

WITH
-- -----------------------------------------
-- 1) Eligible cohort denominators
-- -----------------------------------------
eligible_total AS (
  SELECT COUNT(DISTINCT patient_id) AS total_n
  FROM eligible_patient_cohort
),

eligible_patients_with_age AS (
  SELECT
      a.patient_id,
      (year(current_date) - year(b.patient_yob)) AS age
  FROM eligible_patient_cohort a
  LEFT JOIN (
      SELECT patient_id, patient_yob
      FROM (
          SELECT
              PATIENT_ID AS patient_id,
              PATIENT_YOB AS patient_yob,
              row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
          FROM com_raw.kom_patient_demographics
          WHERE PATIENT_YOB IS NOT NULL
      ) x
      WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
),

eligible_age_buckets AS (
  SELECT
      patient_id,
      CASE
          WHEN age BETWEEN 0  AND 5  THEN '0-5'
          WHEN age BETWEEN 6  AND 11 THEN '6-11'
          WHEN age BETWEEN 12 AND 16 THEN '12-16'
          WHEN age = 17              THEN '17'
          WHEN age >= 18             THEN '18+'
          ELSE NULL
      END AS age_band
  FROM eligible_patients_with_age
),

eligible_age_denoms AS (
  SELECT
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END) AS n_0_5,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END) AS n_6_11,
      COUNT(DISTINCT CASE WHEN age_band = '12-16' THEN patient_id END) AS n_12_16,
      COUNT(DISTINCT CASE WHEN age_band = '17'    THEN patient_id END) AS n_17,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END) AS n_18_plus
  FROM eligible_age_buckets
),

denom AS (
  SELECT *
  FROM eligible_total
  CROSS JOIN eligible_age_denoms
),

-- -----------------------------------------
-- 2) Matched cohort (numerators)
-- -----------------------------------------
cohort AS (
  SELECT *
  FROM cmpa_insights_internal_schema.comorbidity_claims
  WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
    AND matched_code_present = 1
),

base AS (
  SELECT DISTINCT
      patient_id,
      matched_new_bucket,
      matched_subcategory,
      CASE
          WHEN age BETWEEN 0  AND 5  THEN '0-5'
          WHEN age BETWEEN 6  AND 11 THEN '6-11'
          WHEN age BETWEEN 12 AND 16 THEN '12-16'
          WHEN age = 17              THEN '17'
          WHEN age >= 18             THEN '18+'
          ELSE NULL
      END AS age_band
  FROM cohort
),

bucket_totals AS (
  SELECT
      matched_new_bucket,
      'Overall'                                                          AS matched_subcategory,
      COUNT(DISTINCT patient_id)                                         AS overall_n,
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END)  AS age_0_5_n,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END)  AS age_6_11_n,
      COUNT(DISTINCT CASE WHEN age_band = '12-16' THEN patient_id END)  AS age_12_16_n,
      COUNT(DISTINCT CASE WHEN age_band = '17'    THEN patient_id END)  AS age_17_n,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END)  AS age_18_plus_n
  FROM base
  GROUP BY matched_new_bucket
),

bucket_counts AS (
  SELECT
      matched_new_bucket,
      matched_subcategory,
      COUNT(DISTINCT patient_id)                                         AS overall_n,
      COUNT(DISTINCT CASE WHEN age_band = '0-5'   THEN patient_id END)  AS age_0_5_n,
      COUNT(DISTINCT CASE WHEN age_band = '6-11'  THEN patient_id END)  AS age_6_11_n,
      COUNT(DISTINCT CASE WHEN age_band = '12-16' THEN patient_id END)  AS age_12_16_n,
      COUNT(DISTINCT CASE WHEN age_band = '17'    THEN patient_id END)  AS age_17_n,
      COUNT(DISTINCT CASE WHEN age_band = '18+'   THEN patient_id END)  AS age_18_plus_n
  FROM base
  GROUP BY matched_new_bucket, matched_subcategory
),

final_counts AS (
  SELECT * FROM bucket_totals
  UNION ALL
  SELECT * FROM bucket_counts
)

-- -----------------------------------------
-- 3) Final Output
-- -----------------------------------------
SELECT
    f.matched_new_bucket  AS `Clinical Complication`,
    f.matched_subcategory AS `Complication Subcategory`,

    CONCAT(
        f.overall_n, ' (',
        ROUND(100.0 * f.overall_n / NULLIF(d.total_n, 0), 0),
        '%)'
    ) AS `#Patients`,

    CONCAT(
        f.age_0_5_n, ' (',
        ROUND(100.0 * f.age_0_5_n / NULLIF(d.n_0_5, 0), 0),
        '%)'
    ) AS `0-5 years`,

    CONCAT(
        f.age_6_11_n, ' (',
        ROUND(100.0 * f.age_6_11_n / NULLIF(d.n_6_11, 0), 0),
        '%)'
    ) AS `6-11 years`,

    CONCAT(
        f.age_12_16_n, ' (',
        ROUND(100.0 * f.age_12_16_n / NULLIF(d.n_12_16, 0), 0),
        '%)'
    ) AS `12-16 years`,

    CONCAT(
        f.age_17_n, ' (',
        ROUND(100.0 * f.age_17_n / NULLIF(d.n_17, 0), 0),
        '%)'
    ) AS `17 years`,

    CONCAT(
        f.age_18_plus_n, ' (',
        ROUND(100.0 * f.age_18_plus_n / NULLIF(d.n_18_plus, 0), 0),
        '%)'
    ) AS `18+ years`,

    -- 17+ combined
    CONCAT(
        (f.age_17_n + f.age_18_plus_n), ' (',
        ROUND(100.0 * (f.age_17_n + f.age_18_plus_n) / NULLIF((d.n_17 + d.n_18_plus), 0), 0),
        '%)'
    ) AS `17+ years`

FROM final_counts f
CROSS JOIN denom d
ORDER BY
    f.matched_new_bucket,
    CASE WHEN f.matched_subcategory = 'Overall' THEN 0 ELSE 1 END,
    f.matched_subcategory;
```

```sql
-- Output A: Avg and Median age at first incident per comorbidity
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),

eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),

-- First incident per patient per bucket + subcategory
first_incident_subcategory AS (
    SELECT
        patient_id,
        matched_new_bucket,
        matched_subcategory,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket, matched_subcategory
),

-- First incident per patient per bucket only (for Overall row)
-- Logic: earliest date the patient showed ANY subcategory of that bucket
first_incident_bucket AS (
    SELECT
        patient_id,
        matched_new_bucket,
        'Overall'      AS matched_subcategory,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),

-- Stack both together
first_incident_combined AS (
    SELECT * FROM first_incident_subcategory
    UNION ALL
    SELECT * FROM first_incident_bucket
),

-- Join YOB to compute age at first incident
first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        f.matched_subcategory,
        (year(f.first_incident_date) - year(p.patient_yob)) AS age_at_first_incident
    FROM first_incident_combined f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
)

SELECT
    matched_new_bucket                                AS `Clinical Complication`,
    matched_subcategory                               AS `Complication Subcategory`,
    COUNT(DISTINCT patient_id)                        AS `Patient Count`,
    ROUND(AVG(age_at_first_incident), 1)              AS `Avg Age at First Incident`,
    ROUND(PERCENTILE(age_at_first_incident, 0.5), 1)  AS `Median Age at First Incident`
FROM first_incident_with_age
WHERE age_at_first_incident IS NOT NULL
GROUP BY matched_new_bucket, matched_subcategory
ORDER BY
    matched_new_bucket,
    CASE WHEN matched_subcategory = 'Overall' THEN 0 ELSE 1 END,
    matched_subcategory;
```

```sql
-- Output B: Age bucket distribution at first incident (bell curve data)
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),

eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),

first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),

first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        CASE
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 0  AND 5  THEN '0-5'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 6  AND 11 THEN '6-11'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 12 AND 16 THEN '12-16'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) = 17              THEN '17'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) >= 18             THEN '18+'
            ELSE NULL
        END AS age_band_at_incident,
        CASE
            WHEN (year(current_date) - year(p.patient_yob)) BETWEEN 0  AND 5  THEN '0-5'
            WHEN (year(current_date) - year(p.patient_yob)) BETWEEN 6  AND 11 THEN '6-11'
            WHEN (year(current_date) - year(p.patient_yob)) BETWEEN 12 AND 16 THEN '12-16'
            WHEN (year(current_date) - year(p.patient_yob)) = 17               THEN '17'
            WHEN (year(current_date) - year(p.patient_yob)) >= 18              THEN '18+'
            ELSE NULL
        END AS age_band_current
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
),

-- Shared denominator: all patients with a valid incident age band
bucket_totals AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_in_bucket
    FROM first_incident_with_age
    WHERE age_band_at_incident IS NOT NULL
    GROUP BY matched_new_bucket
),

-- Incident age counts per band
age_counts AS (
    SELECT
        matched_new_bucket,
        age_band_at_incident AS age_band,
        COUNT(DISTINCT patient_id) AS n_patients_incident
    FROM first_incident_with_age
    WHERE age_band_at_incident IS NOT NULL
    GROUP BY matched_new_bucket, age_band_at_incident
),

-- Current age counts per band, over the SAME base population (age_band_at_incident IS NOT NULL)
current_age_counts AS (
    SELECT
        matched_new_bucket,
        age_band_current AS age_band,
        COUNT(DISTINCT patient_id) AS n_patients_current
    FROM first_incident_with_age
    WHERE age_band_at_incident IS NOT NULL  -- same base population
      AND age_band_current      IS NOT NULL
    GROUP BY matched_new_bucket, age_band_current
)

SELECT
    a.matched_new_bucket                                                           AS `Clinical Complication`,
    a.age_band                                                                     AS `Age Band`,
    a.n_patients_incident                                                          AS `# Patients (At Incident Age)`,
    ROUND(
        100.0 * a.n_patients_incident / NULLIF(b.total_patients_in_bucket, 0),
        1
    )                                                                              AS `% of Complication Population (Incident Age)`,
    COALESCE(c.n_patients_current, 0)                                              AS `# Patients (Current Age in Same Band)`,
    ROUND(
        100.0 * COALESCE(c.n_patients_current, 0) / NULLIF(b.total_patients_in_bucket, 0),
        1
    )                                                                              AS `% of Complication Population (Current Age)`
FROM age_counts a
JOIN bucket_totals b
    ON a.matched_new_bucket = b.matched_new_bucket
LEFT JOIN current_age_counts c
    ON  a.matched_new_bucket = c.matched_new_bucket
    AND a.age_band           = c.age_band           -- join on same band LABEL, not cross-mapping
ORDER BY
    a.matched_new_bucket,
    CASE
        WHEN a.age_band = '0-5'   THEN 1
        WHEN a.age_band = '6-11'  THEN 2
        WHEN a.age_band = '12-16' THEN 3
        WHEN a.age_band = '17'    THEN 4
        WHEN a.age_band = '18+'   THEN 5
        ELSE 6
    END;
```

```sql
-- Output B: Age distribution at first incident (bell curve data)
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),
eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),
first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),
first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        (year(f.first_incident_date) - year(p.patient_yob)) AS age_at_first_incident,
        (year(current_date) - year(p.patient_yob))          AS current_age
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
),
bucket_totals AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_in_bucket
    FROM first_incident_with_age
    WHERE age_at_first_incident IS NOT NULL
    GROUP BY matched_new_bucket
),

-- Total per bucket based on current age (denominator for current-age %)
current_age_bucket_totals_overall AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_current_age
    FROM first_incident_with_age
    WHERE current_age IS NOT NULL
    GROUP BY matched_new_bucket
),

age_counts AS (
    SELECT
        matched_new_bucket,
        age_at_first_incident AS age,
        COUNT(DISTINCT patient_id) AS n_patients
    FROM first_incident_with_age
    WHERE age_at_first_incident IS NOT NULL
    GROUP BY matched_new_bucket, age_at_first_incident
),

-- Patients whose current age matches a given incident age value
current_age_counts AS (
    SELECT
        matched_new_bucket,
        current_age,
        COUNT(DISTINCT patient_id) AS n_patients_current_age
    FROM first_incident_with_age
    WHERE age_at_first_incident IS NOT NULL
      AND current_age IS NOT NULL
    GROUP BY matched_new_bucket, current_age
)

SELECT
    a.matched_new_bucket                                                        AS `Clinical Complication`,
    a.age                                                                       AS `Age at First Incident`,
    a.n_patients                                                                AS `# Patients`,
    ROUND(100.0 * a.n_patients / NULLIF(b.total_patients_in_bucket, 0), 1)     AS `% of Complication Population`,
    COALESCE(c.n_patients_current_age, 0)                                       AS `# Patients (Current Age in Same Band)`,
    ROUND(
        100.0 * COALESCE(c.n_patients_current_age, 0)
              / NULLIF(d.total_patients_current_age, 0),
        1
    )                                                                           AS `% of Complication Population (Current Age)`
FROM age_counts a
JOIN bucket_totals b
    ON a.matched_new_bucket = b.matched_new_bucket
LEFT JOIN current_age_counts c
    ON  a.matched_new_bucket = c.matched_new_bucket
    AND a.age                = c.current_age
LEFT JOIN current_age_bucket_totals_overall d
    ON  a.matched_new_bucket = d.matched_new_bucket
ORDER BY a.matched_new_bucket, a.age;
```

```sql
-- Output B: Age bucket distribution at first incident (bell curve data)
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),

eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),

first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),

first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        CASE
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 0  AND 5  THEN '0-5'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 6  AND 11 THEN '6-11'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 12 AND 16 THEN '12-16'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) = 17              THEN '17'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) >= 18             THEN '18+'
            ELSE NULL
        END AS age_band
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
),

bucket_totals AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_in_bucket
    FROM first_incident_with_age
    WHERE age_band IS NOT NULL
    GROUP BY matched_new_bucket
),

age_counts AS (
    SELECT
        matched_new_bucket,
        age_band,
        COUNT(DISTINCT patient_id) AS n_patients
    FROM first_incident_with_age
    WHERE age_band IS NOT NULL
    GROUP BY matched_new_bucket, age_band
)

SELECT
    a.matched_new_bucket   AS `Clinical Complication`,
    a.age_band             AS `Age Band at First Incident`,
    a.n_patients           AS `# Patients`,
    ROUND(
        100.0 * a.n_patients / NULLIF(b.total_patients_in_bucket, 0),
        1
    )                      AS `% of Complication Population`
FROM age_counts a
JOIN bucket_totals b ON a.matched_new_bucket = b.matched_new_bucket
ORDER BY
    a.matched_new_bucket,
    CASE
        WHEN a.age_band = '0-5'   THEN 1
        WHEN a.age_band = '6-11'  THEN 2
        WHEN a.age_band = '12-16' THEN 3
        WHEN a.age_band = '17'    THEN 4
        WHEN a.age_band = '18+'   THEN 5
        ELSE 6
    END;
```

```sql
%python
# ============================================================
#  Age Distribution Charts — Databricks Notebook
#  Run each cell separately or as a single cell
# ============================================================

# CELL 1 — Pull data directly from your query
# ============================================================

df = spark.sql("""
WITH
patient_yob AS (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
        SELECT
            PATIENT_ID    AS patient_id,
            PATIENT_YOB   AS patient_yob,
            row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
        FROM com_raw.kom_patient_demographics
        WHERE PATIENT_YOB IS NOT NULL
    ) x
    WHERE rn = 1
),

eligible_claims AS (
    SELECT *
    FROM cmpa_insights_internal_schema.comorbidity_claims
    WHERE patient_id IN (SELECT DISTINCT patient_id FROM eligible_patient_cohort)
      AND matched_code_present = 1
),

first_incident AS (
    SELECT
        patient_id,
        matched_new_bucket,
        MIN(fill_date) AS first_incident_date
    FROM eligible_claims
    GROUP BY patient_id, matched_new_bucket
),

first_incident_with_age AS (
    SELECT
        f.patient_id,
        f.matched_new_bucket,
        CASE
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 0  AND 5  THEN '0-5'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 6  AND 11 THEN '6-11'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) BETWEEN 12 AND 16 THEN '12-16'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) = 17              THEN '17'
            WHEN (year(f.first_incident_date) - year(p.patient_yob)) >= 18             THEN '18+'
            ELSE NULL
        END AS age_band
    FROM first_incident f
    LEFT JOIN patient_yob p ON f.patient_id = p.patient_id
),

bucket_totals AS (
    SELECT
        matched_new_bucket,
        COUNT(DISTINCT patient_id) AS total_patients_in_bucket
    FROM first_incident_with_age
    WHERE age_band IS NOT NULL
    GROUP BY matched_new_bucket
),

age_counts AS (
    SELECT
        matched_new_bucket,
        age_band,
        COUNT(DISTINCT patient_id) AS n_patients
    FROM first_incident_with_age
    WHERE age_band IS NOT NULL
    GROUP BY matched_new_bucket, age_band
)

SELECT
    a.matched_new_bucket   AS clinical_complication,
    a.age_band,
    a.n_patients,
    ROUND(100.0 * a.n_patients / NULLIF(b.total_patients_in_bucket, 0), 1) AS pct
FROM age_counts a
JOIN bucket_totals b ON a.matched_new_bucket = b.matched_new_bucket
ORDER BY
    a.matched_new_bucket,
    CASE
        WHEN a.age_band = '0-5'   THEN 1
        WHEN a.age_band = '6-11'  THEN 2
        WHEN a.age_band = '12-16' THEN 3
        WHEN a.age_band = '17'    THEN 4
        WHEN a.age_band = '18+'   THEN 5
        ELSE 6
    END
""")

# Convert to Pandas for plotting
df_pd = df.toPandas()
display(df_pd)


# CELL 2 — Build the charts
# ============================================================

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# ── Config ──────────────────────────────────────────────────
AGE_BANDS   = ["0-5", "6-11", "12-16", "17", "18+"]
BAR_COLOR   = "#F5A623"
BG_COLOR    = "#F0F0F0"
LABEL_FONT  = 8
TITLE_FONT  = 9
AXIS_FONT   = 8

# ── Pivot data ───────────────────────────────────────────────
conditions = df_pd["clinical_complication"].unique().tolist()

# n per condition (total patients)
n_map = (
    df_pd.groupby("clinical_complication")["n_patients"]
    .sum()
    .to_dict()
)

# pct per condition per age band
pct_map = {}
for condition in conditions:
    sub = df_pd[df_pd["clinical_complication"] == condition]
    pct_map[condition] = dict(zip(sub["age_band"], sub["pct"]))

# ── Layout: 4 top + 3 bottom (last cell hidden) ─────────────
n_conditions = len(conditions)
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
fig.patch.set_facecolor("#FFFFFF")

ax_flat = axes.flatten()
# Hide any unused panels
for i in range(n_conditions, len(ax_flat)):
    ax_flat[i].set_visible(False)

for idx, condition in enumerate(conditions):
    ax  = ax_flat[idx]
    pct = pct_map[condition]
    n   = n_map[condition]

    values = [pct.get(b, 0.0) for b in AGE_BANDS]

    # Panel background & spines
    ax.set_facecolor(BG_COLOR)
    for spine in ax.spines.values():
        spine.set_visible(False)

    bars = ax.bar(AGE_BANDS, values, color=BAR_COLOR, width=0.55, zorder=2)

    # Y-axis
    ax.set_ylim(0, 75)
    ax.yaxis.set_major_locator(mticker.MultipleLocator(10))
    ax.tick_params(axis="y", labelsize=AXIS_FONT, length=0)
    ax.set_ylabel("% of Patients", fontsize=AXIS_FONT, labelpad=4)
    ax.yaxis.grid(True, color="white", linewidth=0.8, zorder=1)
    ax.set_axisbelow(True)

    # X-axis
    ax.tick_params(axis="x", labelsize=AXIS_FONT, length=0)
    ax.set_xlabel("Age Group", fontsize=AXIS_FONT, labelpad=4)

    # Bar value labels
    for bar, val in zip(bars, values):
        if val > 0:
            ax.text(
                bar.get_x() + bar.get_width() / 2,
                bar.get_height() + 0.8,
                f"{val:.0f}%",
                ha="center", va="bottom",
                fontsize=LABEL_FONT, fontweight="bold", color="#333333",
            )

    # Chart title
    ax.set_title(
        f"{condition} (n = {n})",
        fontsize=TITLE_FONT, fontweight="normal", pad=6, color="#222222",
    )

plt.tight_layout(pad=1.5, h_pad=2.5, w_pad=1.5)
plt.savefig("/dbfs/tmp/age_distribution_charts.png", dpi=180, bbox_inches="tight", facecolor="white")
plt.show()
print("✅ Chart saved to /dbfs/tmp/age_distribution_charts.png")


# CELL 3 — (Optional) Save chart to DBFS / your storage path
# ============================================================
# If you want to save to a specific DBFS or mounted path, update below:

# dbutils.fs.cp("dbfs:/tmp/age_distribution_charts.png", "dbfs:/mnt/your-storage/age_distribution_charts.png")
# print("✅ Copied to mounted storage")
```

```sql
CREATE OR REPLACE TABLE cmpa_insights_internal_schema.comorbidity_claims_occurence_data  AS
WITH code_map AS (
  SELECT DISTINCT
    new_bucket,
    subcategory,
    UPPER(TRIM(code)) AS code,
    code_type,
    CASE
      WHEN code_type IN ('ICD-9-CM Diagnosis', 'ICD-10-CM Diagnosis') THEN 'DX'
      WHEN code_type IN ('CPT', 'HCPCS Level II') THEN 'PROC'
      ELSE 'OTHER'
    END AS match_domain
  FROM codes_map_v1
  WHERE subcategory in 
  (
      'ADHD',
      'Anxiety Disorders',
      'Behavioral Issues',
      'Developmental delay',
      'Hydrocephalus',
      'Lack of Physiological Development',
      'Mood disorders',
      'Seizures',
      'Carpal tunnel syndrome',
      'Hearing Aids',
      'Hearing loss',
      'Retinopathy',
      'Sleep related disorders',
      'Spinal Disorders'
)
),

dx_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'DX'
),

proc_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'PROC'
),

medical_base AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
),

medical_dx_exploded AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type,
    REGEXP_REPLACE(UPPER(TRIM(dx_token)), '\\.', '') AS dx_code
  FROM medical_base
  LATERAL VIEW OUTER explode(split(coalesce(diagnosis_codes,''), '\\|')) dx AS dx_token
),

medical_dx AS (
  SELECT
    me.patient_id,
    me.medical_event_id AS claim_id,
    me.service_date      AS fill_date,
    'MEDICAL'            AS source,
    me.diagnosis_codes   AS diagnosis_codes,
    me.procedure_code    AS procedure_code,
    me.procedure_code_type AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM medical_dx_exploded me
  JOIN dx_map dm
    ON me.dx_code = REGEXP_REPLACE(dm.code, '\\.', '')
  WHERE me.dx_code IS NOT NULL AND me.dx_code <> ''
),

medical_proc AS (
  SELECT
    mb.patient_id,
    mb.medical_event_id AS claim_id,
    mb.service_date      AS fill_date,
    'MEDICAL'            AS source,
    mb.diagnosis_codes   AS diagnosis_codes,
    mb.procedure_code    AS procedure_code,
    mb.procedure_code_type AS procedure_code_type,
    'PROC'               AS types,
    pm.code              AS matched_code,
    pm.new_bucket        AS matched_new_bucket,
    pm.subcategory       AS matched_subcategory
  FROM medical_base mb
  JOIN proc_map pm
    ON UPPER(TRIM(mb.procedure_code)) = pm.code
),

pharmacy_base AS (
  SELECT
    patient_id,
    pharmacy_event_id,
    fill_date,
    diagnosis_code
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
    AND transaction_status IN ('Final', 'Standalone')
    AND transaction_result = 'Paid'
),

pharmacy_dx AS (
  SELECT
    pb.patient_id,
    pb.pharmacy_event_id AS claim_id,
    pb.fill_date         AS fill_date,
    'PHARMACY'           AS source,
    pb.diagnosis_code    AS diagnosis_codes,
    CAST(NULL AS STRING) AS procedure_code,
    CAST(NULL AS STRING) AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM pharmacy_base pb
  JOIN dx_map dm
    ON REGEXP_REPLACE(UPPER(TRIM(pb.diagnosis_code)), '\\.', '') = REGEXP_REPLACE(dm.code, '\\.', '')
),

final_union AS (
  SELECT * FROM medical_dx
  UNION ALL
  SELECT * FROM medical_proc
  UNION ALL
  SELECT * FROM pharmacy_dx
),

qc_check AS (
  SELECT
    t.*,
    CASE
      WHEN t.matched_code IS NULL OR TRIM(t.matched_code) = '' THEN 0
      WHEN LOWER(t.diagnosis_codes) LIKE '%|' || LOWER(t.matched_code) || '|%'
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) = LOWER(t.matched_code)
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) LIKE '%' || LOWER(t.matched_code) || '%'
      THEN 1
      ELSE 0
    END AS matched_code_present
  FROM final_union t
),

-- ✅ 1-claim rule: patient must have >= 1 DISTINCT fill_date for the SAME matched_code
-- ✅ CORRECT: Subcategory-level qualification
qualifying_patient_subcategory AS (
  SELECT
    patient_id,
    matched_new_bucket,
    matched_subcategory
  FROM qc_check
  WHERE matched_code_present = 1
  GROUP BY
    patient_id,
    matched_new_bucket,
    matched_subcategory
  HAVING COUNT(DISTINCT fill_date) >= 1   -- 👈 Code B logic
),

filtered_claims AS (
  SELECT q.*
  FROM qc_check q
  JOIN qualifying_patient_subcategory k
    ON q.patient_id = k.patient_id
   AND q.matched_new_bucket = k.matched_new_bucket
   AND q.matched_subcategory = k.matched_subcategory
  WHERE q.matched_code_present = 1
),

patient_age AS (
  SELECT
    a.*,
    (year(current_date) - year(b.patient_yob)) AS age
  FROM filtered_claims a
  LEFT JOIN (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
      SELECT
        PATIENT_ID AS patient_id,
        PATIENT_YOB AS patient_yob,
        row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
      FROM com_raw.kom_patient_demographics
    )
    WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
)

SELECT * FROM patient_age where patient_id in (select distinct patient_id from eligible_patient_cohort);
```

```sql
CREATE OR REPLACE temp view patient_comorbidity_matrix AS
SELECT
    patient_id,

    MAX(CASE WHEN matched_subcategory = 'Spinal Disorders' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS spinal_disorders,
    MAX(CASE WHEN matched_subcategory = 'Developmental delay' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS developmental_delay,
    MAX(CASE WHEN matched_subcategory = 'Hearing loss' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS hearing_loss,
    MAX(CASE WHEN matched_subcategory = 'Lack of Physiological Development' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS lack_of_physiological_development,
    MAX(CASE WHEN matched_subcategory = 'Carpal tunnel syndrome' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS carpal_tunnel_syndrome,
    MAX(CASE WHEN matched_subcategory = 'ADHD' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS adhd,
    MAX(CASE WHEN matched_subcategory = 'Seizures' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS seizures,
    MAX(CASE WHEN matched_subcategory = 'Hearing Aids' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS hearing_aids,
    MAX(CASE WHEN matched_subcategory = 'Mood disorders' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS mood_disorders,
    MAX(CASE WHEN matched_subcategory = 'Hydrocephalus' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS hydrocephalus,
    MAX(CASE WHEN matched_subcategory = 'Anxiety Disorders' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS anxiety_disorders,
    MAX(CASE WHEN matched_subcategory = 'Behavioral Issues' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS behavioral_issues,
    MAX(CASE WHEN matched_subcategory = 'Sleep related disorders' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS sleep_related_disorders,
    MAX(CASE WHEN matched_subcategory = 'Retinopathy' AND fill_date IS NOT NULL THEN 1 ELSE 0 END) AS retinopathy,

    COUNT(DISTINCT CASE
        WHEN fill_date IS NOT NULL THEN matched_subcategory
    END) AS distinct_comorbidity_count

FROM cmpa_insights_internal_schema.comorbidity_claims_occurence_data
GROUP BY patient_id;
```

```sql
WITH patient_subcat_claims AS (
    SELECT
        patient_id,
        matched_subcategory,
        COUNT(DISTINCT fill_date) AS fill_date_claim_cnt
    FROM cmpa_insights_internal_schema.comorbidity_claims_occurence_data
    WHERE matched_subcategory IS NOT NULL
    GROUP BY patient_id, matched_subcategory
),

final_counts AS (
    SELECT
        matched_subcategory,
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt = 1 THEN patient_id END) AS pts_1_cnt,
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt = 2 THEN patient_id END) AS pts_2_cnt,
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt >= 3 THEN patient_id END) AS pts_3_cnt
    FROM patient_subcat_claims
    GROUP BY matched_subcategory
),

cns_counts AS (
    SELECT
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt = 1 THEN patient_id END) AS total_1,
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt = 2 THEN patient_id END) AS total_2,
        COUNT(DISTINCT CASE WHEN fill_date_claim_cnt >= 3 THEN patient_id END) AS total_3
    FROM patient_subcat_claims
),

final_output AS (
    SELECT
        f.matched_subcategory,

        -- COUNT
        f.pts_1_cnt,
        f.pts_2_cnt,
        f.pts_3_cnt,

        -- %
        ROUND(100.0 * f.pts_1_cnt / NULLIF(c.total_1, 0), 1) AS pts_1_pct,
        ROUND(100.0 * f.pts_2_cnt / NULLIF(c.total_2, 0), 1) AS pts_2_pct,
        ROUND(100.0 * f.pts_3_cnt / NULLIF(c.total_3, 0), 1) AS pts_3_pct

    FROM final_counts f
    CROSS JOIN cns_counts c
)

-- CNS TOTAL ROW
SELECT
    'CNS' AS matched_subcategory,
    total_1 AS pts_1_cnt,
    total_2 AS pts_2_cnt,
    total_3 AS pts_3_cnt,
    100.0 AS pts_1_pct,
    100.0 AS pts_2_pct,
    100.0 AS pts_3_pct
FROM cns_counts

UNION ALL

-- SUBCATEGORY ROWS
SELECT
    matched_subcategory,
    pts_1_cnt,
    pts_2_cnt,
    pts_3_cnt,
    pts_1_pct,
    pts_2_pct,
    pts_3_pct
FROM final_output
```

```sql
 %python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Step 1: Load data
# -----------------------------
df = spark.table("cmpa_insights_internal_schema.comorbidity_claims_occurence_data") \
          .select("patient_id", "matched_subcategory") \
          .dropDuplicates() \
          .toPandas()

# -----------------------------
# Step 2: Create patient-subcategory matrix
# -----------------------------
pivot = pd.crosstab(df['patient_id'], df['matched_subcategory'])

# Convert to binary (presence/absence)
pivot = (pivot > 0).astype(int)

# -----------------------------
# Step 3: Co-occurrence matrix
# -----------------------------
co_occurrence = pivot.T.dot(pivot)

# -----------------------------
# Step 4: Order columns (A → Z)
# -----------------------------
categories = sorted(co_occurrence.index.tolist(), key=lambda x: x.lower())

co_occurrence = co_occurrence.loc[categories, categories]

# -----------------------------
# Step 5: Mask upper triangle (KEEP diagonal)
# -----------------------------
mask = np.triu(np.ones_like(co_occurrence, dtype=bool), k=1)

co_occurrence_masked = co_occurrence.mask(mask)

# -----------------------------
# Step 6: Plot Heatmap
# -----------------------------
plt.figure(figsize=(12, 10))

plt.imshow(
    co_occurrence_masked,
    cmap='RdYlGn',
    vmin=0,
    vmax=co_occurrence.values.max()
)

plt.xticks(range(len(categories)), categories, rotation=90)
plt.yticks(range(len(categories)), categories)

# Move x-axis labels to top
plt.gca().xaxis.tick_top()
plt.gca().xaxis.set_label_position('top')

# -----------------------------
# Step 7: Add values (INCLUDING FULL DIAGONAL)
# -----------------------------
for i in range(len(categories)):
    for j in range(len(categories)):
        if not mask[i, j]:
            val = int(co_occurrence.iloc[i, j])

            # Bold diagonal (all single-condition counts)
            if i == j:
                plt.text(
                    j, i, val,
                    ha="center", va="center",
                    fontsize=8, fontweight='bold'
                )
            else:
                plt.text(
                    j, i, val,
                    ha="center", va="center",
                    fontsize=8
                )

# -----------------------------
# Step 8: Title + Colorbar
# -----------------------------
plt.title(
    "Comorbidity Co-occurrence (Counts)\n(Diagonal = Total Patients per Condition)",
    pad=40
)

cbar = plt.colorbar(label='Patient Count')
ticks = np.linspace(0, co_occurrence.values.max(), 6).astype(int)
cbar.set_ticks(ticks)

plt.tight_layout()
plt.show()
```

```sql
-- =============================================================================
-- PATIENT LEVEL COMORBIDITY MATRIX + DISTINCT COUNT (NO BUCKETS)
-- =============================================================================

CREATE OR REPLACE TABLE cmpa_insights_internal_schema.comorbidity_patient_summary AS

WITH base AS (
    SELECT DISTINCT
        patient_id,
        age,
        matched_subcategory
    FROM cmpa_insights_internal_schema.comorbidity_claims_occurence_data
),

-- --------------------------------------------------
-- Pivot to wide format (1/0 flags)
-- --------------------------------------------------
pivoted AS (
    SELECT
        patient_id,
        age,
        MAX(CASE WHEN matched_subcategory = 'ADHD' THEN 1 ELSE 0 END) AS adhd,
        MAX(CASE WHEN matched_subcategory = 'Anxiety Disorders' THEN 1 ELSE 0 END) AS anxiety_disorders,
        MAX(CASE WHEN matched_subcategory = 'Behavioral Issues' THEN 1 ELSE 0 END) AS behavioral_issues,
        MAX(CASE WHEN matched_subcategory = 'Developmental delay' THEN 1 ELSE 0 END) AS developmental_delay,
        MAX(CASE WHEN matched_subcategory = 'Hydrocephalus' THEN 1 ELSE 0 END) AS hydrocephalus,
        MAX(CASE WHEN matched_subcategory = 'Lack of Physiological Development' THEN 1 ELSE 0 END) AS lack_of_physiological_development,
        MAX(CASE WHEN matched_subcategory = 'Mood disorders' THEN 1 ELSE 0 END) AS mood_disorders,
        MAX(CASE WHEN matched_subcategory = 'Seizures' THEN 1 ELSE 0 END) AS seizures,
        MAX(CASE WHEN matched_subcategory = 'Carpal tunnel syndrome' THEN 1 ELSE 0 END) AS carpal_tunnel_syndrome,
        MAX(CASE WHEN matched_subcategory = 'Hearing Aids' THEN 1 ELSE 0 END) AS hearing_aids,
        MAX(CASE WHEN matched_subcategory = 'Hearing loss' THEN 1 ELSE 0 END) AS hearing_loss,
        MAX(CASE WHEN matched_subcategory = 'Retinopathy' THEN 1 ELSE 0 END) AS retinopathy,
        MAX(CASE WHEN matched_subcategory = 'Sleep related disorders' THEN 1 ELSE 0 END) AS sleep_related_disorders,
        MAX(CASE WHEN matched_subcategory = 'Spinal Disorders' THEN 1 ELSE 0 END) AS spinal_disorders

    FROM base
    GROUP BY patient_id, age
)

-- --------------------------------------------------
-- Final output with distinct count
-- --------------------------------------------------
SELECT
    *,
    (
        adhd +
        anxiety_disorders +
        behavioral_issues +
        developmental_delay +
        hydrocephalus +
        lack_of_physiological_development +
        mood_disorders +
        seizures +
        carpal_tunnel_syndrome +
        hearing_aids +
        hearing_loss +
        retinopathy +
        sleep_related_disorders +
        spinal_disorders
    ) AS distinct_comorbidities

FROM pivoted;
```

```sql
WITH base AS (
    SELECT
        patient_id,
        CASE
            WHEN distinct_comorbidities = 1 THEN '1'
            WHEN distinct_comorbidities = 2 THEN '2'
            WHEN distinct_comorbidities = 3 THEN '3'
            WHEN distinct_comorbidities = 4 THEN '4'
            WHEN distinct_comorbidities = 5 THEN '5'
            WHEN distinct_comorbidities = 6 THEN '6'
            WHEN distinct_comorbidities = 7 THEN '7'
            WHEN distinct_comorbidities = 8 THEN '8'
            WHEN distinct_comorbidities = 9 THEN '9'
            WHEN distinct_comorbidities = 10 THEN '10'
            WHEN distinct_comorbidities >=11 THEN '11+'
        END AS bucket
    FROM cmpa_insights_internal_schema.comorbidity_patient_summary
),

total AS (
    SELECT COUNT(DISTINCT patient_id) AS total_n FROM base
)

SELECT
    b.bucket,
    COUNT(DISTINCT b.patient_id) AS patients,
    ROUND(100.0 * COUNT(DISTINCT b.patient_id) / t.total_n, 1) AS pct

FROM base b
CROSS JOIN total t
GROUP BY b.bucket, t.total_n
ORDER BY b.bucket;
```

```sql
CREATE OR REPLACE TABLE cmpa_insights_internal_schema.comorbidity_claims_occurence_data_2dxclaims  AS
WITH code_map AS (
  SELECT DISTINCT
    new_bucket,
    subcategory,
    UPPER(TRIM(code)) AS code,
    code_type,
    CASE
      WHEN code_type IN ('ICD-9-CM Diagnosis', 'ICD-10-CM Diagnosis') THEN 'DX'
      WHEN code_type IN ('CPT', 'HCPCS Level II') THEN 'PROC'
      ELSE 'OTHER'
    END AS match_domain
  FROM codes_map_v1
  WHERE subcategory in 
  (
      'ADHD',
      'Anxiety Disorders',
      'Behavioral Issues',
      'Developmental delay',
      'Hydrocephalus',
      'Lack of Physiological Development',
      'Mood disorders',
      'Seizures',
      'Carpal tunnel syndrome',
      'Hearing Aids',
      'Hearing loss',
      'Retinopathy',
      'Sleep related disorders',
      'Spinal Disorders'
)
),

dx_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'DX'
),

proc_map AS (
  SELECT /*+ BROADCAST */ *
  FROM code_map
  WHERE match_domain = 'PROC'
),

medical_base AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type
  FROM com_edp_prd.com_raw.kom_medical_events
  WHERE service_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
),

medical_dx_exploded AS (
  SELECT
    patient_id,
    medical_event_id,
    service_date,
    diagnosis_codes,
    procedure_code,
    procedure_code_type,
    REGEXP_REPLACE(UPPER(TRIM(dx_token)), '\\.', '') AS dx_code
  FROM medical_base
  LATERAL VIEW OUTER explode(split(coalesce(diagnosis_codes,''), '\\|')) dx AS dx_token
),

medical_dx AS (
  SELECT
    me.patient_id,
    me.medical_event_id AS claim_id,
    me.service_date      AS fill_date,
    'MEDICAL'            AS source,
    me.diagnosis_codes   AS diagnosis_codes,
    me.procedure_code    AS procedure_code,
    me.procedure_code_type AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM medical_dx_exploded me
  JOIN dx_map dm
    ON me.dx_code = REGEXP_REPLACE(dm.code, '\\.', '')
  WHERE me.dx_code IS NOT NULL AND me.dx_code <> ''
),

medical_proc AS (
  SELECT
    mb.patient_id,
    mb.medical_event_id AS claim_id,
    mb.service_date      AS fill_date,
    'MEDICAL'            AS source,
    mb.diagnosis_codes   AS diagnosis_codes,
    mb.procedure_code    AS procedure_code,
    mb.procedure_code_type AS procedure_code_type,
    'PROC'               AS types,
    pm.code              AS matched_code,
    pm.new_bucket        AS matched_new_bucket,
    pm.subcategory       AS matched_subcategory
  FROM medical_base mb
  JOIN proc_map pm
    ON UPPER(TRIM(mb.procedure_code)) = pm.code
),

pharmacy_base AS (
  SELECT
    patient_id,
    pharmacy_event_id,
    fill_date,
    diagnosis_code
  FROM com_edp_prd.com_raw.kom_pharmacy_events
  WHERE fill_date BETWEEN TO_DATE('${start_date}') AND TO_DATE('${end_date}')
    AND transaction_status IN ('Final', 'Standalone')
    AND transaction_result = 'Paid'
),

pharmacy_dx AS (
  SELECT
    pb.patient_id,
    pb.pharmacy_event_id AS claim_id,
    pb.fill_date         AS fill_date,
    'PHARMACY'           AS source,
    pb.diagnosis_code    AS diagnosis_codes,
    CAST(NULL AS STRING) AS procedure_code,
    CAST(NULL AS STRING) AS procedure_code_type,
    'DX'                 AS types,
    dm.code              AS matched_code,
    dm.new_bucket        AS matched_new_bucket,
    dm.subcategory       AS matched_subcategory
  FROM pharmacy_base pb
  JOIN dx_map dm
    ON REGEXP_REPLACE(UPPER(TRIM(pb.diagnosis_code)), '\\.', '') = REGEXP_REPLACE(dm.code, '\\.', '')
),

final_union AS (
  SELECT * FROM medical_dx
  UNION ALL
  SELECT * FROM medical_proc
  UNION ALL
  SELECT * FROM pharmacy_dx
),

qc_check AS (
  SELECT
    t.*,
    CASE
      WHEN t.matched_code IS NULL OR TRIM(t.matched_code) = '' THEN 0
      WHEN LOWER(t.diagnosis_codes) LIKE '%|' || LOWER(t.matched_code) || '|%'
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) = LOWER(t.matched_code)
        OR LOWER(CAST(t.procedure_code AS VARCHAR(255))) LIKE '%' || LOWER(t.matched_code) || '%'
      THEN 1
      ELSE 0
    END AS matched_code_present
  FROM final_union t
),

-- ✅ 1-claim rule: patient must have >= 2 DISTINCT fill_date for the SAME matched_code
-- ✅ CORRECT: Subcategory-level qualification
qualifying_patient_subcategory AS (
  SELECT
    patient_id,
    matched_new_bucket,
    matched_subcategory
  FROM qc_check
  WHERE matched_code_present = 1
  GROUP BY
    patient_id,
    matched_new_bucket,
    matched_subcategory
  HAVING COUNT(DISTINCT fill_date) >= 2
),

filtered_claims AS (
  SELECT q.*
  FROM qc_check q
  JOIN qualifying_patient_subcategory k
    ON q.patient_id = k.patient_id
   AND q.matched_new_bucket = k.matched_new_bucket
   AND q.matched_subcategory = k.matched_subcategory
  WHERE q.matched_code_present = 1
),

patient_age AS (
  SELECT
    a.*,
    (year(current_date) - year(b.patient_yob)) AS age
  FROM filtered_claims a
  LEFT JOIN (
    SELECT DISTINCT patient_id, patient_yob
    FROM (
      SELECT
        PATIENT_ID AS patient_id,
        PATIENT_YOB AS patient_yob,
        row_number() OVER (PARTITION BY PATIENT_ID ORDER BY PATIENT_YOB DESC) AS rn
      FROM com_raw.kom_patient_demographics
    )
    WHERE rn = 1
  ) b
    ON a.patient_id = b.patient_id
)

SELECT * FROM patient_age where patient_id in (select distinct patient_id from eligible_patient_cohort);
```

```sql
WITH eligible AS (
  SELECT DISTINCT patient_id
  FROM eligible_patient_cohort
),
patient_flags AS (
  SELECT
    e.patient_id,

    /* unique patient-level flags */
    MAX(CASE
          WHEN c.matched_code_present = 1
           AND UPPER(c.matched_new_bucket) = 'NEUROLOGICAL DISORDERS'
           AND UPPER(c.matched_subcategory) = 'HEARING LOSS'
          THEN 1 ELSE 0
        END) AS has_hearing_loss,

    MAX(CASE
          WHEN c.matched_code_present = 1
           AND UPPER(c.matched_new_bucket) = 'NEUROLOGICAL DISORDERS'
           AND UPPER(c.matched_subcategory) = 'HEARING AIDS'
          THEN 1 ELSE 0
        END) AS has_hearing_aids

  FROM eligible e
  LEFT JOIN cmpa_insights_internal_schema.comorbidity_claims c
    ON e.patient_id = c.patient_id
  GROUP BY e.patient_id
)

SELECT
  /* these are all unique patients because patient_flags is 1 row per patient */
  COUNT(*) AS eligible_unique_patients,

  COUNT(CASE WHEN has_hearing_loss = 1 AND has_hearing_aids = 0 THEN 1 END) AS hearing_loss_only_unique_patients,
  COUNT(CASE WHEN has_hearing_loss = 0 AND has_hearing_aids = 1 THEN 1 END) AS hearing_aids_only_unique_patients,
  COUNT(CASE WHEN has_hearing_loss = 1 AND has_hearing_aids = 1 THEN 1 END) AS hearing_loss_and_aids_unique_patients,
  COUNT(CASE WHEN has_hearing_loss = 0 AND has_hearing_aids = 0 THEN 1 END) AS neither_unique_patients
FROM patient_flags;
```

```sql
WITH eligible AS (
  SELECT DISTINCT patient_id
  FROM eligible_patient_cohort
),
patient_flags AS (
  SELECT
    e.patient_id,

    MAX(CASE
          WHEN c.matched_code_present = 1
           AND UPPER(c.matched_new_bucket) = 'NEUROLOGICAL DISORDERS'
           AND UPPER(c.matched_subcategory) = 'HEARING LOSS'
          THEN 1 ELSE 0
        END) AS hearing_loss_flag,

    MAX(CASE
          WHEN c.matched_code_present = 1
           AND UPPER(c.matched_new_bucket) = 'NEUROLOGICAL DISORDERS'
           AND UPPER(c.matched_subcategory) = 'HEARING AIDS'
          THEN 1 ELSE 0
        END) AS hearing_aids_flag

  FROM eligible e
  LEFT JOIN cmpa_insights_internal_schema.comorbidity_claims c
    ON e.patient_id = c.patient_id
  GROUP BY e.patient_id
)

SELECT
  patient_id,
  hearing_loss_flag,
  hearing_aids_flag,

  CASE WHEN hearing_loss_flag = 1 AND hearing_aids_flag = 1 THEN 1 ELSE 0 END AS both_flag,
  CASE WHEN hearing_loss_flag = 1 AND hearing_aids_flag = 0 THEN 1 ELSE 0 END AS hearing_loss_only_flag,
  CASE WHEN hearing_loss_flag = 0 AND hearing_aids_flag = 1 THEN 1 ELSE 0 END AS hearing_aids_only_flag

FROM patient_flags;
```

```sql
-- =============================================================================
-- Continuous enrollment check for ONLY: index_date - 6 months  to  index_date + 6 months
-- Business rules SAME as your current logic:
--   • merge overlapping/adjacent enrollment rows (<= 1 day gap allowed)
--   • pick the merged block that covers index_date
--   • flag = 1 if that block fully covers [index-6mo, index+6mo]
-- Output: distinct patient_id with continuous coverage in +/- 6 months window
-- =============================================================================

create or replace temporary view continuous_enrollment_patients_pm6m as
WITH idx AS (
  SELECT
    patient_id,
    first_elaprase_treatment_date AS index_date,

    -- window bounds: +/- 6 months
    add_months(first_elaprase_treatment_date, -6) AS pre_6m_start,
    add_months(first_elaprase_treatment_date,  6) AS post_6m_end

  FROM first_elaprase_treatment
),

-- only enrollment rows that could affect the +/- 6 month window
enr AS (
  SELECT
    e.patient_id,
    e.start_date,
    e.end_date
  FROM com_raw.kom_patient_enrollment e
  JOIN idx i
    ON e.patient_id = i.patient_id
  WHERE e.end_date  >= i.pre_6m_start
    AND e.start_date <= i.post_6m_end
),

-- detect gaps (gaps & islands) with 1-day adjacency allowed (same rule as above)
ordered AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    CASE
      WHEN lag(end_date) OVER (
             PARTITION BY patient_id ORDER BY start_date, end_date
           ) IS NULL
      THEN 1
      WHEN start_date <= date_add(
             lag(end_date) OVER (PARTITION BY patient_id ORDER BY start_date, end_date),
             1
           )
      THEN 0
      ELSE 1
    END AS new_block
  FROM enr
),

blocks AS (
  SELECT
    patient_id,
    start_date,
    end_date,
    sum(new_block) OVER (
      PARTITION BY patient_id ORDER BY start_date, end_date
      ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS block_id
  FROM ordered
),

merged AS (
  SELECT
    patient_id,
    block_id,
    min(start_date) AS block_start,
    max(end_date)   AS block_end
  FROM blocks
  GROUP BY patient_id, block_id
),

-- get the block that covers index_date
covering AS (
  SELECT
    i.patient_id,
    i.index_date,
    i.pre_6m_start,
    i.post_6m_end,
    m.block_start,
    m.block_end
  FROM idx i
  LEFT JOIN merged m
    ON i.patient_id = m.patient_id
   AND i.index_date BETWEEN m.block_start AND m.block_end
),

flags AS (
  SELECT
    patient_id,
    index_date,
    CASE
      WHEN block_start <= pre_6m_start
       AND block_end   >= post_6m_end
      THEN 1 ELSE 0
    END AS cont_pm_6m_flag
  FROM covering
)

SELECT DISTINCT patient_id
FROM flags
WHERE cont_pm_6m_flag = 1;
```

```sql
SELECT 
    CASE 
        WHEN PATIENT_AGE BETWEEN 0 AND 5 THEN '0-5'
        WHEN PATIENT_AGE BETWEEN 6 AND 11 THEN '6-11'
        WHEN PATIENT_AGE BETWEEN 12 AND 16 THEN '12-16'
        WHEN PATIENT_AGE = 17 THEN '17'
        WHEN PATIENT_AGE >= 18 THEN '18+'
        ELSE 'Unknown'
    END AS age_band,
    COUNT(DISTINCT patient_id) AS patient_count
FROM com_edp_prd.cmpa_insights_internal_schema.patient360_master
GROUP BY 
    CASE 
        WHEN PATIENT_AGE BETWEEN 0 AND 5 THEN '0-5'
        WHEN PATIENT_AGE BETWEEN 6 AND 11 THEN '6-11'
        WHEN PATIENT_AGE BETWEEN 12 AND 16 THEN '12-16'
        WHEN PATIENT_AGE = 17 THEN '17'
        WHEN PATIENT_AGE >= 18 THEN '18+'
        ELSE 'Unknown'
    END
ORDER BY age_band;
```
